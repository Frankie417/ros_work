## 导入依赖

```python
import math
import numpy as np
from arm5dof_uservo import Arm5DoFUServo
from config import *
```

## 创建机械臂

```python
arm = Arm5DoFUServo()
```

## 末端控制(自动生成Pitch)

```python
p_tool = [150.0, 80.0, 200.0]
arm.set_tool_pose(p_tool)
```

```python
p_tool = [200.0, -40.0, 100.0]
arm.set_tool_pose(p_tool)
```

```python
p_tool =  [140.0, 0.0, 150.0]
tool_roll = np.radians(-135.0)
arm.set_tool_pose(p_tool, tool_roll=tool_roll)
```

```python
p_tool = [180.0, -50.0, 100.0]
tool_roll = np.radians(90.0)
arm.set_tool_pose(p_tool, tool_roll=tool_roll)
```

## 末端控制(设定Pitch)

```python
tool_posi = [180.0, -50.0, 100.0]
tool_pitch = np.radians(110.0)
tool_roll = np.radians(180.0)
arm.set_tool_pose(tool_posi, tool_pitch=tool_pitch, tool_roll=tool_roll)
```

## 获取当前末端位姿

```python
# 返回的数据格式: [x6, y6, z6, pitch, roll]  
x, y, z, pitch, roll = arm.get_tool_pose()
print(f"坐标: X {x:.1f}  Y {y:.1f} Z {z:.1f}")
print(f"俯仰角(Pitch) {np.degrees(pitch):.1f} deg")
print(f"横滚角(Roll) {np.degrees(roll):.1f} deg")
```
