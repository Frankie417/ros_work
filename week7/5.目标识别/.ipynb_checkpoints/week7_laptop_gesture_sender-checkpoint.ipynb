# 实验七：电脑本地手势识别与 Jetson 指令发送

**必须把这个 notebook 下载到自己的电脑，在电脑本地启动 Jupyter 并选择电脑本地 Python 内核。**在浏览器中打开 Jetson 的 Jupyter 时，代码仍在 Jetson 上运行，不能使用电脑摄像头。

先在电脑环境安装 `mediapipe`、`opencv-python`、`ipywidgets`、`numpy`，例如在终端运行 `python -m pip install mediapipe opencv-python ipywidgets numpy jupyterlab`。第 1 个代码单元下载官方预训练模型。无需训练，也不使用 Jetson 摄像头。

手势：One→选择张开夹爪；Two→选择闭合夹爪；Three→选择回到零点；OK→确认发送；Five→停止锁定。Three/OK 的分数是关键点几何余量，不是校准的概率。

```
from pathlib import Path
from urllib.request import urlretrieve

目录 = Path.cwd()
模型路径 = 目录 / 'gesture_recognizer.task'
模型网址 = 'https://storage.googleapis.com/mediapipe-models/gesture_recognizer/gesture_recognizer/float16/1/gesture_recognizer.task'
if not 模型路径.is_file():
    print('正在下载官方预训练模型到：', 模型路径)
    urlretrieve(模型网址, 模型路径)
print('模型已就绪：', 模型路径)
```

```
import time
from pathlib import Path
import cv2
import numpy as np
import mediapipe as mp
import math

目录 = Path.cwd()
模型路径 = 目录 / 'gesture_recognizer.task'
相机号 = 0
最低置信度 = 0.70
手势中文 = {'One': '食指竖起', 'Two': '食指和中指竖起', 'Three': '食指、中指和无名指竖起', 'OK': '拇指与食指成环', 'Five': '五指张开'}
有效手势 = tuple(手势中文)
伸直阈值 = 1.85
弯曲阈值 = 1.55
OK距离阈值 = 0.42

def 创建识别器(模式):
    """用已下载的官方模型创建 MediaPipe 识别器。"""
    if not 模型路径.is_file():
        raise FileNotFoundError('缺少手势模型：' + str(模型路径))
    选项 = mp.tasks.vision.GestureRecognizerOptions(
        base_options=mp.tasks.BaseOptions(model_asset_path=str(模型路径)),
        running_mode=模式,
        num_hands=1,
        min_hand_detection_confidence=0.5,
        min_hand_presence_confidence=0.5,
        min_tracking_confidence=0.5,
    )
    return mp.tasks.vision.GestureRecognizer.create_from_options(选项)

def 识别画面(图像, 识别器, 时间戳=None):
    """用官方类别识别一指、二指和五指张开，用关键点判断三指与 OK。"""
    彩色 = cv2.cvtColor(图像, cv2.COLOR_BGR2RGB)
    输入图 = mp.Image(image_format=mp.ImageFormat.SRGB, data=np.ascontiguousarray(彩色))
    if 时间戳 is None:
        原始结果 = 识别器.recognize(输入图)
    else:
        原始结果 = 识别器.recognize_for_video(输入图, 时间戳)
    官方类别, 官方分数 = '无手', 0.0
    if 原始结果.gestures and 原始结果.gestures[0]:
        首位 = 原始结果.gestures[0][0]
        官方类别, 官方分数 = 首位.category_name, float(首位.score)
    类别, 分数, 分数来源 = '未定义手势', 0.0, '无'
    if not 原始结果.hand_landmarks:
        类别 = '无手'
    else:
        点 = 原始结果.hand_landmarks[0]
        宽高比 = 图像.shape[1] / 图像.shape[0]
        def 距离(甲, 乙):
            return math.hypot((点[甲].x - 点[乙].x) * 宽高比, 点[甲].y - 点[乙].y)
        比值 = [距离(尖端, 根部) / max(距离(近端, 根部), 1e-6) for 根部, 近端, 尖端 in ((5,6,8),(9,10,12),(13,14,16),(17,18,20))]
        环距 = 距离(4, 8) / max(距离(5, 17), 1e-6)
        if 环距 <= OK距离阈值 and all(比值[i] >= 伸直阈值 for i in (1,2,3)):
            余量 = min(OK距离阈值-环距, *(比值[i]-伸直阈值 for i in (1,2,3)))
            类别, 分数, 分数来源 = 'OK', min(0.99, 0.70 + max(0.0, 余量)), '关键点几何余量'
        elif all(比值[i] >= 伸直阈值 for i in (0,1,2)) and 比值[3] <= 弯曲阈值:
            余量 = min(*(比值[i]-伸直阈值 for i in (0,1,2)), 弯曲阈值-比值[3])
            类别, 分数, 分数来源 = 'Three', min(0.99, 0.70 + max(0.0, 余量)), '关键点几何余量'
        elif 官方类别 in ('Pointing_Up', 'Victory', 'Open_Palm') and (官方类别 != 'Open_Palm' or 环距 > OK距离阈值):
            类别 = {'Pointing_Up':'One', 'Victory':'Two', 'Open_Palm':'Five'}[官方类别]
            分数, 分数来源 = 官方分数, '模型分类分数'
    通过 = 类别 in 有效手势 and 分数 >= 最低置信度 and len(原始结果.hand_landmarks) == 1
    return {'类别': 类别, '置信度': 分数, '通过': 通过, '分数来源': 分数来源, '官方类别': 官方类别, '官方分数': 官方分数, '关键点': 原始结果.hand_landmarks}

def 标注画面(图像, 结果):
    """在真实画面上绘制模型手部关键点和类别。"""
    标注 = 图像.copy()
    高, 宽 = 标注.shape[:2]
    手部连线 = ((0,1),(1,2),(2,3),(3,4),(0,5),(5,6),(6,7),(7,8),
              (5,9),(9,10),(10,11),(11,12),(9,13),(13,14),(14,15),(15,16),
              (13,17),(0,17),(17,18),(18,19),(19,20))
    for 手 in 结果['关键点']:
        for 点序号甲, 点序号乙 in 手部连线:
            甲, 乙 = 手[点序号甲], 手[点序号乙]
            cv2.line(标注, (int(甲.x*宽), int(甲.y*高)), (int(乙.x*宽), int(乙.y*高)), (0, 255, 0), 2)
        for 点 in 手:
            cv2.circle(标注, (int(点.x*宽), int(点.y*高)), 3, (0, 0, 255), -1)
    文字 = '{} {:.2f}'.format(结果['类别'], 结果['置信度'])
    cv2.putText(标注, 文字, (12, 28), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
    return 标注

print('模型文件已找到：{}；可识别手势：{}'.format(模型路径.name, '、'.join(手势中文[x] for x in 有效手势)))
```

```
from collections import deque
import threading

连续确认帧数 = 3
五指张开停止帧数 = 2
候选超时秒 = 8.0
动作映射 = {
    'One': '张开夹爪',
    'Two': '闭合夹爪',
    'Three': '回到零点',
}
确认手势 = 'OK'
停止手势 = 'Five'

class 手势状态机:
    """连续帧选动作、OK 确认、五指张开锁定停止，执行期间互锁。"""
    def __init__(self):
        self.状态 = 'WAIT'
        self.已准备 = False
        self.忙碌 = False
        self.停止锁定 = False
        self.候选 = None
        self.候选时间 = 0.0
        self.历史 = deque(maxlen=连续确认帧数)
        self.五指张开计数 = 0
        self.上次执行手势 = None

    def 准备(self):
        if self.忙碌 or self.停止锁定:
            return False
        self.状态, self.已准备, self.候选 = 'WAIT', True, None
        self.历史.clear()
        return True

    def 停止(self):
        self.停止锁定 = True
        self.已准备 = False
        self.候选 = None
        self.历史.clear()
        self.状态 = 'STOPPED'

    def 解除停止(self):
        if self.忙碌:
            return False
        self.停止锁定 = False
        self.状态 = 'WAIT'
        self.五指张开计数 = 0
        self.历史.clear()
        return True

    def 输入(self, 类别, 分数, 时间=None):
        """只返回一次待执行动作；五指张开优先进入停止锁定。"""
        时间 = time.monotonic() if 时间 is None else 时间
        if self.停止锁定:
            return None
        有效 = 类别 if 分数 >= 最低置信度 else None
        self.五指张开计数 = self.五指张开计数 + 1 if 有效 == 停止手势 else 0
        if self.五指张开计数 >= 五指张开停止帧数:
            self.停止()
            return '停止锁定'
        if not self.已准备 or self.忙碌 or self.停止锁定:
            return None
        if self.状态 == 'SELECT' and 时间 - self.候选时间 > 候选超时秒:
            self.状态, self.候选 = 'WAIT', None
            self.历史.clear()
        允许 = 有效 if 有效 in (*动作映射, 确认手势) else None
        self.历史.append(允许)
        if len(self.历史) < 连续确认帧数 or len(set(self.历史)) != 1 or 允许 is None:
            return None
        self.历史.clear()
        if self.状态 == 'WAIT' and 允许 in 动作映射:
            self.状态, self.候选, self.候选时间 = 'SELECT', 允许, 时间
            return None
        if self.状态 == 'SELECT' and 允许 == 确认手势:
            self.状态 = 'CONFIRM'
            动作 = 动作映射[self.候选]
            self.上次执行手势 = self.候选
            self.状态, self.忙碌, self.已准备 = 'EXECUTE', True, False
            return 动作
        if self.状态 == 'SELECT' and 允许 in 动作映射:
            self.候选, self.候选时间 = 允许, 时间
        return None

    def 执行结束(self):
        self.忙碌 = False
        self.状态 = 'STOPPED' if self.停止锁定 else 'WAIT'
        self.候选 = None

状态机 = 手势状态机()
print('已设定选择、确认、停止与执行互锁。')
```

## 实操

1. Jetson 的接收 notebook 点击“启动接收服务”，复制本次连接令牌。
2. 在下方填地址与令牌，点击“测试连接”，确认显示“演练”。
3. 点击“启动电脑摄像头”，检查画面上五种英文标签。点击“准备下一条指令”，稳定展示 One/Two/Three 中的一种，手放松，再做 OK。每次动作只发一次。
4. Five 连续识别会发送停止锁定；点击“解除停止锁定”恢复。手动“发送停止锁定”也可锁定。**这不能中断已经开始的动作。**
5. 演练完成后，在 Jetson 接收 notebook 手动点击“连接并启用真机”，再逐项测试。记录本轮测试和保存截图只记录实际现场发生的结果，演练和真机模式分别标明。

```
import csv
import json
import threading
from datetime import datetime
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError
import ipywidgets as widgets
from IPython.display import display

证据目录 = 目录 / 'week7_laptop_evidence'
记录路径 = 目录 / 'week7_laptop_trials.csv'
命令代码 = {'张开夹爪': 'open_gripper', '闭合夹爪': 'close_gripper', '回到零点': 'home'}
停止事件 = threading.Event()
取帧线程 = None
最近原图 = None
最近结果 = None
最近动作 = '未触发'
本轮已记录 = False

地址框 = widgets.Text(value='http://192.168.55.1:8765', description='Jetson 地址', layout=widgets.Layout(width='420px'))
令牌框 = widgets.Password(description='连接令牌', layout=widgets.Layout(width='420px'))
测试按钮 = widgets.Button(description='测试连接')
启动按钮 = widgets.Button(description='启动电脑摄像头')
停止按钮 = widgets.Button(description='停止并释放摄像头')
准备按钮 = widgets.Button(description='准备下一条指令')
解除按钮 = widgets.Button(description='解除停止锁定')
发送停止按钮 = widgets.Button(description='发送停止锁定')
截图按钮 = widgets.Button(description='保存截图')
记录按钮 = widgets.Button(description='记录本轮测试')
预期框 = widgets.Dropdown(options=[(手势中文[x] + '→' + 动作映射[x], x) for x in 动作映射] + [('无动作', '无动作')], description='预期指令')
备注框 = widgets.Text(description='现场备注')
状态框 = widgets.HTML(value='请先启动 Jetson 接收端，填令牌并测试连接。')
画面框 = widgets.Image(format='jpeg', width=640)

def 请求(路径, 指令=None):
    地址 = 地址框.value.rstrip('/') + 路径
    请求头 = {'X-Week7-Token': 令牌框.value}
    数据 = None
    if 指令 is not None:
        请求头['Content-Type'] = 'application/json'
        数据 = json.dumps({'command': 指令}).encode('utf-8')
    对象 = Request(地址, data=数据, headers=请求头, method='POST' if 数据 is not None else 'GET')
    try:
        with urlopen(对象, timeout=3) as 响应:
            return json.loads(响应.read().decode('utf-8'))
    except HTTPError as 错误:
        try:
            内容 = json.loads(错误.read().decode('utf-8'))
        except (ValueError, UnicodeDecodeError):
            内容 = {'错误': str(错误)}
        raise RuntimeError(内容.get('错误', str(错误))) from 错误
    except URLError as 错误:
        raise RuntimeError('无法连接 Jetson：' + str(错误.reason)) from 错误

def 测试连接(_):
    try:
        远端 = 请求('/status')
        状态框.value = '连接正常；Jetson 模式：{}；忙碌：{}；停止锁定：{}'.format(远端['模式'], 远端['忙碌'], 远端['停止锁定'])
    except Exception as 错误:
        状态框.value = '连接失败：' + str(错误)

def 发送动作(动作):
    global 最近动作
    try:
        回复 = 请求('/command', 命令代码[动作])
        最近动作 = 'Jetson 已接受：{}（{}）'.format(回复['动作'], 回复['模式'])
        状态框.value = 最近动作 + '；等待远端动作结束后才能准备下一条。'
    except Exception as 错误:
        最近动作 = '发送失败或结果不明：' + str(错误)
        状态框.value = 最近动作 + '；请先查询 Jetson 状态，勿直接重发。'
    finally:
        状态机.执行结束()

def 发送停止():
    global 最近动作
    状态机.停止()
    try:
        回复 = 请求('/command', 'stop')
        最近动作 = 'Jetson 停止锁定：' + str(回复['停止锁定'])
    except Exception as 错误:
        最近动作 = '远端停止未确认：' + str(错误)
    状态框.value = 最近动作

def 取帧循环():
    global 最近原图, 最近结果, 最近动作
    摄像头 = cv2.VideoCapture(相机号)
    识别器 = None
    try:
        if not 摄像头.isOpened():
            raise RuntimeError('电脑摄像头无法打开；确认此 notebook 使用电脑本地内核，并关闭其他占用程序。')
        识别器 = 创建识别器(mp.tasks.vision.RunningMode.VIDEO)
        上次时间戳 = -1
        while not 停止事件.is_set():
            成功, 图像 = 摄像头.read()
            if not 成功:
                raise RuntimeError('电脑摄像头取帧失败。')
            时间戳 = max(int(time.monotonic()*1000), 上次时间戳+1)
            上次时间戳 = 时间戳
            结果 = 识别画面(图像, 识别器, 时间戳)
            最近原图, 最近结果 = 图像.copy(), 结果
            事件 = 状态机.输入(结果['类别'], 结果['置信度'])
            if 事件 == '停止锁定':
                threading.Thread(target=发送停止, daemon=True).start()
            elif 事件 in 命令代码:
                threading.Thread(target=发送动作, args=(事件,), daemon=True).start()
            成功编码, 编码 = cv2.imencode('.jpg', 标注画面(图像, 结果))
            if 成功编码:
                画面框.value = 编码.tobytes()
            状态框.value = '状态：{}；候选：{}；识别：{} {:.2f}；最近动作：{}'.format(
                状态机.状态, 手势中文.get(状态机.候选, '无'),
                手势中文.get(结果['类别'], 结果['类别']), 结果['置信度'], 最近动作)
            停止事件.wait(0.12)
    except Exception as 错误:
        状态框.value = '实时识别出错：' + str(错误)
    finally:
        if 识别器 is not None:
            识别器.close()
        摄像头.release()
        状态机.停止()

def 启动(_):
    global 取帧线程
    if 取帧线程 is not None and 取帧线程.is_alive():
        状态框.value = '摄像头已运行。'
        return
    try:
        请求('/status')
    except Exception as 错误:
        状态框.value = '先连接 Jetson 接收端：' + str(错误)
        return
    停止事件.clear()
    取帧线程 = threading.Thread(target=取帧循环, daemon=True)
    取帧线程.start()

def 停止(_):
    状态机.停止()
    停止事件.set()
    状态框.value = '摄像头正在释放；这不会发送远端停止指令。'

def 准备(_):
    global 最近动作, 本轮已记录
    try:
        远端 = 请求('/status')
        if 远端['忙碌'] or 远端['停止锁定']:
            状态框.value = 'Jetson 忙碌或停止锁定；请先处理远端状态。'
            return
        if 状态机.准备():
            最近动作, 本轮已记录 = '未触发', False
            状态框.value = '已准备；做 One、Two 或 Three，放松，再做 OK 确认。'
        else:
            状态框.value = '电脑本地仍忙碌或停止锁定。'
    except Exception as 错误:
        状态框.value = '查询 Jetson 失败：' + str(错误)

def 解除(_):
    try:
        请求('/command', 'clear_stop')
        if 状态机.解除停止():
            状态框.value = '远端和本地均已解除停止锁定；点击准备后再发指令。'
    except Exception as 错误:
        状态框.value = '解除失败：' + str(错误)

def 停止指令(_):
    threading.Thread(target=发送停止, daemon=True).start()

def 截图(_):
    if 最近原图 is None:
        状态框.value = '尚无电脑摄像头真实画面。'
        return
    证据目录.mkdir(exist_ok=True)
    名称 = datetime.now().strftime('%Y%m%d_%H%M%S_%f') + '_gesture.png'
    cv2.imwrite(str(证据目录 / 名称), 标注画面(最近原图, 最近结果))
    状态框.value = '真实截图已保存：' + str(证据目录 / 名称)

def 记录(_):
    global 本轮已记录
    if 最近原图 is None or 本轮已记录 or 状态机.忙碌:
        状态框.value = '需有真实画面、动作发送结束，且本轮尚未记录。'
        return
    try:
        远端 = 请求('/status')
    except Exception as 错误:
        状态框.value = '查询 Jetson 失败，暂不记录：' + str(错误)
        return
    if 远端['忙碌']:
        状态框.value = 'Jetson 动作尚未完成，请稍后记录。'
        return
    证据目录.mkdir(exist_ok=True)
    时间 = datetime.now().strftime('%Y%m%d_%H%M%S_%f')
    名称 = 时间 + '_trial.png'
    cv2.imwrite(str(证据目录 / 名称), 标注画面(最近原图, 最近结果))
    预期 = 动作映射.get(预期框.value, '无动作')
    实际 = 远端['最后动作']
    结果 = 远端['最后结果']
    正确 = int(最近动作.startswith('Jetson 已接受：' + 预期) and 实际 == 预期 and 结果 in ('真机动作已返回', '演练已接收')) if 预期 != '无动作' else int(最近动作 == '未触发')
    新文件 = not 记录路径.exists()
    with 记录路径.open('a', newline='', encoding='utf-8-sig') as 文件:
        表格 = csv.writer(文件)
        if 新文件:
            表格.writerow(['时间', '预期手势', '预期动作', '当前识别手势', '识别分数', 'Jetson 最后动作', 'Jetson 最后结果', '是否正确', '模式', '现场备注', '截图'])
        表格.writerow([时间, 预期框.value, 预期, 最近结果['类别'], '{:.4f}'.format(最近结果['置信度']), 实际, 结果, 正确, 远端['模式'], 备注框.value, 名称])
    本轮已记录 = True
    状态框.value = '本轮真实测试已记录：' + str(记录路径)

测试按钮.on_click(测试连接)
启动按钮.on_click(启动)
停止按钮.on_click(停止)
准备按钮.on_click(准备)
解除按钮.on_click(解除)
发送停止按钮.on_click(停止指令)
截图按钮.on_click(截图)
记录按钮.on_click(记录)
display(widgets.VBox([
    widgets.HBox([地址框, 令牌框, 测试按钮]),
    widgets.HBox([启动按钮, 停止按钮, 准备按钮, 解除按钮, 发送停止按钮]),
    画面框, widgets.HBox([预期框, 备注框, 截图按钮, 记录按钮]), 状态框]))
```
