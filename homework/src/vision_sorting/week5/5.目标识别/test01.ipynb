```python
import cv2
import time
import matplotlib.pyplot as plt

cap = cv2.VideoCapture(0, cv2.CAP_V4L2)

print("摄像头打开状态:", cap.isOpened())

try:
    if cap.isOpened():
        time.sleep(0.5)

        for i in range(5):
            ret, frame = cap.read()
            print(f"第 {i+1} 次读取: {ret}")

            if ret and frame is not None:
                rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

                plt.imshow(rgb)
                plt.axis("off")
                plt.show()
                break

            time.sleep(0.2)
finally:
    cap.release()
```
