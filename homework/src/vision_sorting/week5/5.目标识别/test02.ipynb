```python
import cv2
import matplotlib.pyplot as plt

cap = cv2.VideoCapture(0, cv2.CAP_V4L2)

print("摄像头打开:", cap.isOpened())

try:
    if cap.isOpened():
        ret, frame = cap.read()
        print("读取结果:", ret)

        if ret:
            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            plt.imshow(rgb)
            plt.axis("off")
            plt.show()
finally:
    cap.release()
```
