```python
import time
import numpy as np
from numpy import pi
import pybullet
import pybullet_data
# 自定义库
from arm5dof_pybullet import Arm5DoFPyBullet

# 设置Numpy的打印选项
# 精确位数3，不启用科学计数法
np.set_printoptions(precision=5, suppress=True)
```

## 创建机械臂

```python
# 创建机械臂
# 传入机械臂路径
arm = Arm5DoFPyBullet("urdf/arm5dof.urdf")
```

## 关节状态查询

```python
# 获取单个关节角度
angle = arm.get_joint_angle("joint1")
print(f"joint1 angle {angle:.1f}")
```

```python
# 获取多个关节角度
theta_list = arm.get_joint_angle_list()
# 打印关节角度列表
print(theta_list)
```

```python
# 获取单个关节的速度
velocity = arm.get_joint_velocity("joint1")
print(f"joint1 velocity {velocity:.4f}")
```

```python
# 批量获取多个关节的速度
velocity_list = arm.get_joint_velocity_list()
print(velocity_list)
```

## 关节角度控制

```python
# 设置单个关节的角度
arm.set_joint_angle("joint1", -pi/4, is_wait=True)
```

```python
# 设置多个关节的角度
# 依次传入 关节1，关节2，关节3，关节4， 夹爪的角度
theta_list = np.radians([0.0, -90.0, 90.0, -50.0, 25.0, 45])
arm.set_joint_angle_list(theta_list, is_wait=True)
```

```python
arm.gripper_close()
```

## 工具末端姿态查询

```python
# 获取工具坐标系末端位姿
xyz_pitch = arm.get_tool_pose()
tool_posi = xyz_pitch[:-2]
tool_pitch, tool_roll = xyz_pitch[-2:]

print(f"末端位置: {tool_posi}  俯仰角(RY): {tool_pitch} 横滚角(RZ2): {tool_roll}")
```

## 逆向运动学测试

```python
# 目标位姿
tool_posi = [250, 0.0, 200] # 工具坐标系位置 单位mm
tool_pitch = pi/2 # 工具坐标系俯仰角 pitch, ZYZ欧拉角
tool_roll = pi
# 逆向运动学
ret, target_joint_angle = arm.inverse_kinematic(tool_posi, tool_pitch, is_debug=True)
print(f"是否有解: {ret}")
print(f"目标关节弧度: {target_joint_angle}")
```

## 设置工具末端姿态

```python
# 设置工具坐标系末端位姿
tool_posi = [70, 200, 150] # 工具坐标系位置 单位mm
arm.set_tool_pose(tool_posi, T=1.0)
```

```python
# 设置工具坐标系末端位姿
tool_posi = [70, 200, 250] # 工具坐标系位置 单位mm
roll = pi + pi/4
arm.set_tool_pose(tool_posi, tool_roll=roll, T=1.0)
```

```python
# 设置工具坐标系末端位姿
tool_posi = [50, 200, 100] # 工具坐标系位置 单位mm
pitch = pi/2 + pi/8 # 工具坐标系俯仰角 pitch, ZYZ欧拉角
roll = pi
arm.set_tool_pose(tool_posi, pitch, roll, T=2.0)
```

```python
# 设置工具坐标系末端位姿
tool_posi = [50, -200, 200] # 工具坐标系位置 单位mm
arm.set_tool_pose(tool_posi, T=2.0)
```

## 夹爪控制

```python
# 夹爪张开
arm.gripper_open()
```

```python
# 夹爪闭合
arm.gripper_close()
```

## 机械臂停止

```python
# 机械臂是否停止
arm.is_stop()
```

```python
# 机械臂终止运动
arm.stop()
```
