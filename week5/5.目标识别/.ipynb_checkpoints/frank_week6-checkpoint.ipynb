## 2 sorting_core

以下是内置实现，直接运行，不需另存为文件。


```python
"""Fixed-P4 HSV detection, state transitions and deterministic motion plans."""
import copy
import json
import math
from pathlib import Path
import time
import cv2
import numpy as np
COLORS = ('red', 'yellow', 'green', 'blue')
BGR = {'red': (0, 0, 255), 'yellow': (0, 255, 255), 'green': (0, 255, 0), 'blue': (255, 0, 0)}

def check_pose(pose, label):
    if not isinstance(pose, dict):
        raise ValueError(label + ': 缺少位姿')
    values = [pose.get(k) for k in ('x', 'y', 'z', 'pitch_deg', 'roll_deg')]
    if any((isinstance(v, bool) or not isinstance(v, (float, int)) or (not math.isfinite(v)) for v in values)):
        raise ValueError(label + ': 请填写有限的 XYZ/Pitch/Roll，不能使用空值')
    if not -180 <= pose['pitch_deg'] <= 180 or not -180 <= pose['roll_deg'] <= 180:
        raise ValueError(label + ': 角度必须是单圈有效角度，原始异常 Roll 不可直接使用')
    if not 10 <= pose['z'] <= 350:
        raise ValueError(label + ': Z 超出 10..350 mm 检查范围')
    return pose

def make_plan(config, color):
    """Explicit endpoints; never synthesize a +55 mm bin waypoint blindly."""
    if color not in COLORS:
        raise ValueError('未知颜色')
    pick = config['p4']['pick']
    above = config['p4']['above']
    dest = config['bin_map'][color]['pose']
    dest_above = config['bin_map'][color]['above']
    home = config['observation_pose']
    return [dict(op='open', stage='OPEN'), dict(op='move', stage='APPROACH', pose=copy.deepcopy(above)), dict(op='move', stage='PICK', pose=copy.deepcopy(pick)), dict(op='close', stage='GRIP'), dict(op='move', stage='LIFT', pose=copy.deepcopy(above)), dict(op='move', stage='TRANSFER_HOME', pose=copy.deepcopy(home)), dict(op='move', stage='BIN_ABOVE', pose=copy.deepcopy(dest_above)), dict(op='move', stage='PLACE', pose=copy.deepcopy(dest)), dict(op='release', stage='RELEASE'), dict(op='move', stage='RETREAT', pose=copy.deepcopy(dest_above)), dict(op='move', stage='RESET', pose=copy.deepcopy(home))]

def validate_motion_config(config, require_confirmed=True):
    if require_confirmed and (not config.get('physical_calibration_confirmed', False)):
        raise ValueError('P4 低位、桶位 Roll、上方点及路径尚未现场确认；先演练/示教，再设置 physical_calibration_confirmed')
    for color in COLORS:
        for step in make_plan(config, color):
            if step['op'] == 'move':
                check_pose(step['pose'], color + '/' + step['stage'])
    for label, low, high in [('P4', config['p4']['pick'], config['p4']['above'])] + [(c, config['bin_map'][c]['pose'], config['bin_map'][c]['above']) for c in COLORS]:
        if high['z'] <= low['z']:
            raise ValueError(label + ': 上方点必须高于低位')
        if high['x'] != low['x'] or high['y'] != low['y']:
            raise ValueError(label + ': Approach/Retreat 必须同 X/Y 垂直进出')
    for key in ('move_seconds', 'gripper_seconds'):
        if not isinstance(config.get(key), (int, float)) or not math.isfinite(config[key]) or config[key] < 0.3:
            raise ValueError(key + ': 必须是至少 0.3 秒的有限数值')
    for key in ('open_angle_rad', 'release_angle_rad'):
        if not isinstance(config.get(key), (int, float)) or not 0 < config[key] <= math.pi / 2:
            raise ValueError(key + ': 夹爪开口必须在 0..pi/2 弧度内')

def detect_color(frame, config, roi=None):
    """All four colors at once. Return one target, empty, or ambiguous."""
    height, width = frame.shape[:2]
    if roi is None:
        x, y, w, h = (0, 0, width, height)
    else:
        if len(roi) != 4 or not all((isinstance(v, int) for v in roi)):
            raise ValueError('ROI 格式为整数 x,y,width,height')
        x, y, w, h = roi
        if x < 0 or y < 0 or w <= 0 or (h <= 0) or (x + w > width) or (y + h > height):
            raise ValueError('P4 ROI 超出真实画面，请重新设置')
    cropped = frame[y:y + h, x:x + w]
    hsv = cv2.cvtColor(cropped, cv2.COLOR_BGR2HSV)
    kernel = np.ones((config['vision']['kernel_size'],) * 2, np.uint8)
    masks, objects = ({}, [])
    annotated = frame.copy()
    for color in COLORS:
        mask = np.zeros((h, w), np.uint8)
        for lower, upper in config['vision']['hsv'][color]:
            mask = cv2.bitwise_or(mask, cv2.inRange(hsv, np.asarray(lower, np.uint8), np.asarray(upper, np.uint8)))
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
        masks[color] = mask
        for contour in cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)[-2]:
            area = float(cv2.contourArea(contour))
            if area < config['vision']['min_area']:
                continue
            bx, by, bw, bh = cv2.boundingRect(contour)
            moments = cv2.moments(contour)
            center = (int(moments['m10'] / moments['m00']) + x, int(moments['m01'] / moments['m00']) + y)
            obj = dict(color=color, center=center, area=area, bbox=(bx + x, by + y, bw, bh))
            objects.append(obj)
            cv2.rectangle(annotated, (bx + x, by + y), (bx + x + bw, by + y + bh), BGR[color], 2)
            cv2.putText(annotated, color, (bx + x, max(20, by + y - 5)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, BGR[color], 2)
    cv2.rectangle(annotated, (x, y), (x + w - 1, y + h - 1), (255, 255, 255), 2)
    cv2.putText(annotated, 'P4 ROI' if roi else 'PREVIEW - set P4 ROI', (x + 5, y + 24), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 2)
    invalid = float(cv2.cvtColor(cropped, cv2.COLOR_BGR2GRAY).mean()) < config['vision']['min_scene_brightness']
    kind = 'invalid' if invalid else 'target' if len(objects) == 1 else 'empty' if not objects else 'ambiguous'
    return dict(kind=kind, target=objects[0] if kind == 'target' else None, objects=objects, masks=masks, annotated=annotated)

class SortGate:

    def __init__(self, frames=6, empty_frames=8, tolerance=18, reset_delay=0.5):
        self.frames, self.empty_frames, self.tolerance, self.reset_delay = (frames, empty_frames, tolerance, reset_delay)
        self.enabled = False
        self.busy = False
        self.need_empty = False
        self.error = False
        self.state = 'STOPPED'
        self.count = self.empty_count = 0
        self.anchor = None
        self.after = 0

    def start(self):
        if self.busy or self.error:
            raise RuntimeError('动作未结束或存在故障，请先处理')
        self.enabled = True
        self.state = 'RESET' if self.need_empty else 'READY'
        self.count = 0
        self.anchor = None

    def stop(self):
        self.enabled = False
        self.count = 0
        self.anchor = None
        self.state = 'STOPPING' if self.busy else 'STOPPED'

    def feed(self, result, now=None):
        now = time.monotonic() if now is None else now
        if not self.enabled or self.busy or self.error:
            return None
        if self.need_empty:
            self.state = 'RESET'
            if now < self.after:
                return None
            self.empty_count = self.empty_count + 1 if result['kind'] == 'empty' else 0
            if self.empty_count >= self.empty_frames:
                self.need_empty = False
                self.empty_count = 0
                self.state = 'READY'
            return None
        if result['kind'] != 'target':
            self.count = 0
            self.anchor = None
            self.state = 'DETECT'
            return None
        target = result['target']
        same = self.anchor is not None and target['color'] == self.anchor['color'] and (math.hypot(target['center'][0] - self.anchor['center'][0], target['center'][1] - self.anchor['center'][1]) <= self.tolerance)
        self.count = self.count + 1 if same else 1
        if not same:
            self.anchor = copy.deepcopy(target)
        self.state = 'CONFIRM'
        if self.count >= self.frames:
            self.busy = True
            self.state = 'BUSY'
            self.count = 0
            return copy.deepcopy(target)
        return None

    def finish(self, success, now=None):
        self.busy = False
        self.need_empty = True
        self.empty_count = self.count = 0
        self.anchor = None
        self.after = (time.monotonic() if now is None else now) + self.reset_delay
        if not success:
            self.enabled = False
            self.error = True
            self.state = 'ERROR'
        else:
            self.state = 'RESET' if self.enabled else 'STOPPED'
```

## 3 arm_interrupt

以下是内置实现，直接运行，不需另存为文件。


```python
"""Cooperative trajectory cancellation plus feedback-based position hold.

Only the motion worker touches the SDK/UART. The HTTP stop handler sets an Event.
This is a software protective stop, not a certified hardware emergency stop.
"""
import math
from pathlib import Path
import sys
import threading
import time

class Cancelled(Exception):
    pass

class InterruptibleArm:

    def __init__(self, sdk_dir, port, stop_event):
        sys.path.insert(0, str(Path(sdk_dir).resolve()))
        from arm5dof_uservo import Arm5DoFUServo
        from uservo import Packet
        self.stop_event = stop_event
        self.arm = Arm5DoFUServo.__new__(Arm5DoFUServo)
        try:
            Arm5DoFUServo.__init__(self.arm, device=port, is_init_pose=False, config_folder=str(Path(sdk_dir).resolve() / 'config'))
        except (Exception, SystemExit) as error:
            uart = getattr(self.arm, 'uart', None)
            if uart is not None:
                uart.close()
            raise RuntimeError('机械臂初始化失败；检查端口、舵机在线和配置：' + str(error)) from error
        self.manager = self.arm.uservo
        self.ids = sorted(self.manager.servos)
        if self.ids != list(range(6)):
            self.arm.uart.close()
            raise RuntimeError('Expected six online servos with IDs 0..5; found ' + str(self.ids))
        self.allow_hold = False
        self.feedback = {identifier: 0 for identifier in self.ids}
        for code in (self.manager.CODE_QUERY_SERVO_ANGLE, self.manager.CODE_QUERY_SERVO_ANGLE_MTURN):
            original = self.manager.response_handle_funcs[code]

            def response(data, original=original):
                original(data)
                self.feedback[data[0]] = self.feedback.get(data[0], 0) + 1
            self.manager.response_handle_funcs[code] = response
        motion_codes = {getattr(self.manager, name) for name in dir(self.manager) if name.startswith('CODE_SET_SERVO_ANGLE')}

        def send(code, data):
            if code in motion_codes and (not self.allow_hold):
                self.check()
            packet = Packet.pack(code, data)
            if self.arm.uart.write(packet) != len(packet):
                raise IOError('Incomplete servo UART write')
        self.manager.send_request = send
        self.arm.uart.write_timeout = 0.2
        self.original_query = self.manager.query_servo_angle
        self.manager.query_servo_angle = self.query
        self.arm.trajectory_callback_period = 1

    def check(self):
        if self.stop_event.is_set():
            raise Cancelled('Motion cancelled')

    def query(self, identifier):
        previous = self.feedback[identifier]
        angle = self.original_query(identifier)
        if self.feedback[identifier] == previous or angle is None or (not math.isfinite(float(angle))):
            raise IOError('Missing fresh angle feedback from servo ' + str(identifier))
        return float(angle)

    def execute(self, command):
        self.check()
        if command == 'open_gripper':
            self.arm.gripper_open(callback=self.check)
        elif command == 'close_gripper':
            self.arm.gripper_close(callback=self.check)
        elif command == 'home':
            result = self.arm.set_tool_pose(pose_name='home', T=2.0, is_soft_move=True, callback=self.check)
            if result is False:
                raise RuntimeError('Home inverse kinematics failed')
        else:
            raise ValueError('Unknown action')
        deadline = time.monotonic() + 5
        while time.monotonic() < deadline:
            self.check()
            if self.manager.is_stop():
                return
            self.stop_event.wait(0.02)
        raise TimeoutError('Servo did not reach target')

    def hold(self):
        """Override the unfinished target with each servo's freshly measured angle."""
        self.allow_hold = True
        try:
            errors = []
            targets = {}
            for identifier in self.ids:
                try:
                    angle = self.query(identifier)
                    targets[identifier] = angle
                    if self.manager.set_servo_angle(identifier, angle, interval=0, power=0) is False:
                        raise IOError('Servo rejected hold')
                except Exception as error:
                    errors.append(str(error))
            if errors:
                raise RuntimeError('; '.join(errors))
            deadline = time.monotonic() + 1.5
            stable = 0
            previous = {identifier: self.query(identifier) for identifier in self.ids}
            while time.monotonic() < deadline:
                time.sleep(0.05)
                current = {identifier: self.query(identifier) for identifier in self.ids}
                near_target = all((abs(current[i] - targets[i]) <= 1.0 for i in self.ids))
                stable = stable + 1 if near_target and all((abs(current[i] - previous[i]) <= 0.5 for i in self.ids)) else 0
                if stable >= 2:
                    return {'hold_sent': True, 'feedback_stable': True}
                previous = current
            raise RuntimeError('Hold sent but stationary feedback not confirmed')
        finally:
            self.allow_hold = False

    def close(self):
        self.arm.uart.close()

class SimulatedArm:

    def __init__(self, stop_event, duration=2):
        self.stop_event, self.duration = (stop_event, duration)
        self.holds = 0

    def execute(self, command):
        if self.stop_event.wait(self.duration):
            raise Cancelled('Simulation interrupted')

    def hold(self):
        self.holds += 1
        return {'hold_sent': False, 'feedback_stable': False, 'simulation_interrupted': True}

    def close(self):
        pass
```

## 4 hardware_backend

以下是内置实现，直接运行，不需另存为文件。


```python
"""Offline IK before UART connection; one motion worker owns the arm."""
import math
from pathlib import Path
import sys
import time
import numpy as np

def precheck(config, sdk_dir, require_confirmed=True):
    validate_motion_config(config, require_confirmed)
    sdk = Path(sdk_dir).resolve()
    if not (sdk / 'arm5dof_kinematic.py').is_file():
        raise FileNotFoundError('找不到 SDK: ' + str(sdk))
    sys.path.insert(0, str(sdk))
    from arm5dof_kinematic import Arm5DoFKinematic
    k, b = np.loadtxt(str(sdk / 'config/z_error_polyfit.txt'), delimiter=',')
    solver = Arm5DoFKinematic()
    checked = []
    seen = set()
    for color in COLORS:
        for step in make_plan(config, color):
            if step['op'] != 'move':
                continue
            pose = step['pose']
            key = tuple((pose[k] for k in ('x', 'y', 'z', 'pitch_deg', 'roll_deg')))
            if key in seen:
                continue
            seen.add(key)
            z = pose['z'] + float(k) * math.hypot(pose['x'], pose['y']) + float(b)
            solutions = solver.inverse_kinematic(pose['x'], pose['y'], z, math.radians(pose['pitch_deg']), math.radians(pose['roll_deg']))
            if not solutions:
                raise ValueError('IK 不可达，尚未连接舵机：' + color + '/' + step['stage'] + ' ' + str(pose))
            checked.append(dict(color=color, stage=step['stage'], pose=pose, solutions=len(solutions)))
    return checked

class Backend:

    def __init__(self, config, cancel, hardware=False):
        self.config, self.cancel, self.hardware = (config, cancel, hardware)
        self.robot = None
        if hardware:
            precheck(config, config['sdk_dir'])
            self.robot = InterruptibleArm(config['sdk_dir'], config['arm_port'], cancel)

    def check(self):
        if self.cancel.is_set():
            raise Cancelled('分拣已取消')

    def move(self, pose):
        self.check()
        if not self.hardware:
            if self.cancel.wait(0.12):
                self.check()
            return
        arm = self.robot.arm
        adjusted = arm.adjust_tool_posi([pose['x'], pose['y'], pose['z']])
        reachable, _ = arm.inverse_kinematic(adjusted, math.radians(pose['pitch_deg']), math.radians(pose['roll_deg']))
        self.check()
        if not reachable:
            raise RuntimeError('运行时 IK 不可达，停止后续动作')
        result = arm.set_tool_pose([pose['x'], pose['y'], pose['z']], tool_pitch=math.radians(pose['pitch_deg']), tool_roll=math.radians(pose['roll_deg']), T=self.config['move_seconds'], is_soft_move=True, is_wait=True, callback=self.check)
        if result is False:
            raise RuntimeError('SDK 拒绝目标位姿')
        self.wait_settled()

    def wait_settled(self):
        if not self.hardware:
            return
        deadline = time.monotonic() + 5
        while time.monotonic() < deadline:
            self.check()
            if self.robot.manager.is_stop():
                return
            self.cancel.wait(0.02)
        raise TimeoutError('关节未在等待时间内到位')

    def execute(self, color, stage_callback):
        for step in make_plan(self.config, color):
            self.check()
            stage_callback(step['stage'])
            if step['op'] == 'move':
                self.move(step['pose'])
            elif not self.hardware:
                if self.cancel.wait(0.12):
                    self.check()
            else:
                arm = self.robot.arm
                if step['op'] == 'close':
                    arm.gripper_close(T=self.config['gripper_seconds'], callback=self.check)
                else:
                    key = 'release_angle_rad' if step['op'] == 'release' else 'open_angle_rad'
                    arm.gripper_open(angle=self.config[key], T=self.config['gripper_seconds'], callback=self.check)
                self.wait_settled()

    def hold(self):
        if self.hardware:
            return self.robot.hold()
        return {'simulation_interrupted': True}

    def close(self):
        if self.robot:
            self.robot.close()
```

## 5 week6_color_sorting

以下是内置实现，直接运行，不需另存为文件。


```python
"""Jetson Jupyter panel: fixed P4 color sorting, no automatic movement on import."""
import argparse
import copy
import csv
from datetime import datetime
import json
from pathlib import Path
import queue
import threading
import time
import traceback
import cv2
import numpy as np
HERE = Path.cwd()

class Runtime:

    def __init__(self, config, notify=lambda message: None):
        self.config = copy.deepcopy(config)
        self.notify = notify
        self.lock = threading.RLock()
        self.cancel = threading.Event()
        self.camera_stop = threading.Event()
        self.gate = SortGate(frames=config['vision']['confirm_frames'], empty_frames=config['vision']['empty_frames'], tolerance=config['vision']['position_tolerance_px'], reset_delay=config['vision']['reset_delay_seconds'])
        self.backend = Backend(self.config, self.cancel, False)
        self.jobs = queue.Queue()
        self.output = HERE / 'results' / datetime.now().strftime('%Y%m%d_%H%M%S_%f')
        self.image_callback = lambda image: None
        self.last = None
        self.last_time = 0
        self.camera_thread = None
        self.observation_ready = True
        self.roi = None
        self.latest_trial = None
        self.stage = ''
        self.closing = False
        self.actor = threading.Thread(target=self.motion_worker, daemon=True)
        self.actor.start()

    def log(self, event, **fields):
        self.output.mkdir(parents=True, exist_ok=True)
        with self.lock:
            with (self.output / 'events.jsonl').open('a', encoding='utf-8') as stream:
                stream.write(json.dumps(dict(time=datetime.now().isoformat(), event=event, **fields), ensure_ascii=False) + '\n')

    def start_camera(self):
        if self.camera_thread and self.camera_thread.is_alive():
            return
        self.camera_stop.clear()
        self.camera_thread = threading.Thread(target=self.vision_worker, daemon=True)
        self.camera_thread.start()

    def apply_roi(self, roi):
        with self.lock:
            if self.gate.busy:
                raise RuntimeError('动作期间不能更改 P4 ROI')
            if self.last is None:
                raise RuntimeError('先打开摄像头')
            detect_color(self.last['raw'], self.config, roi)
            self.roi = list(roi)
            self.gate.stop()
            self.gate.anchor = None
            self.gate.count = 0
        self.notify('已设置 P4 ROI；只放一个物块，点击开始自动分拣。')

    def start_auto(self):
        with self.lock:
            if self.last is None or time.monotonic() - self.last_time > 1:
                raise RuntimeError('需要新鲜的摄像头画面')
            if self.roi is None:
                raise RuntimeError('先设置只包含 P4 抓取区的 ROI，排除彩色垃圾桶和机械臂')
            if not self.observation_ready:
                raise RuntimeError('先点击回观察位，再重新设置 P4 ROI')
            if self.cancel.is_set():
                raise RuntimeError('已停止或故障；点击解除停止后再开始')
            self.gate.start()
        self.notify('自动分拣已开启；连续稳定识别后执行，不需要手动选择颜色。')

    def stop(self):
        self.cancel.set()
        with self.lock:
            self.gate.stop()
            if not self.gate.busy:
                self.gate.busy = True
                self.gate.state = 'STOPPING'
                self.jobs.put(('hold', None))
        self.notify('已请求中断，等待同一动作线程完成位置保持。')

    def clear_stop(self):
        with self.lock:
            if self.gate.busy or self.gate.error:
                raise RuntimeError('动作/制停未完成或存在故障，先检查日志并重新连接')
            self.cancel.clear()
            self.gate.state = 'STOPPED'
        self.notify('停止已解除；不会自动续接原轨迹，需重新开始自动分拣。')

    def connect(self, hardware):
        with self.lock:
            if self.gate.busy or self.gate.enabled:
                raise RuntimeError('先停止自动分拣并等待动作/制停结束')
            self.gate.busy = True
            self.gate.error = False
            self.gate.state = 'CONNECTING'
            self.cancel.clear()
            self.jobs.put(('connect', bool(hardware)))

    def observe(self):
        with self.lock:
            if self.gate.busy or self.gate.enabled or self.gate.error:
                raise RuntimeError('先停用自动任务并处理故障')
            if self.cancel.is_set():
                raise RuntimeError('先解除停止')
            self.gate.busy = True
            self.gate.state = 'MOVING_OBSERVE'
            self.jobs.put(('observe', None))

    def save_image(self, image, name):
        self.output.mkdir(parents=True, exist_ok=True)
        ok, encoded = cv2.imencode('.png', image)
        if not ok:
            raise IOError('截图编码失败')
        encoded.tofile(str(self.output / name))
        return name

    def snapshot(self):
        with self.lock:
            if self.last is None or time.monotonic() - self.last_time > 1:
                raise RuntimeError('没有新鲜画面')
            image = self.last['annotated'].copy()
        name = datetime.now().strftime('%H%M%S_%f') + '_preview.png'
        self.save_image(image, name)
        self.notify('已保存 ' + str(self.output / name))

    def record_trial(self, actual, actual_bin, verdict, note):
        with self.lock:
            trial = copy.deepcopy(self.latest_trial)
            if not trial or self.gate.busy:
                raise RuntimeError('请在一轮分拣完成后记录')
            path = self.output / 'manual_trials.csv'
            existing = []
            if path.exists():
                with path.open(encoding='utf-8-sig', newline='') as stream:
                    existing = list(csv.DictReader(stream))
            if any((r['trial_id'] == trial['trial_id'] for r in existing)):
                raise RuntimeError('本轮已经记录，请勿重复统计')
            row = dict(trial_id=trial['trial_id'], time=datetime.now().isoformat(), mode=trial['mode'], actual_color=actual, detected_color=trial['color'], expected_bin=self.config['bin_map'][actual]['label'], actual_bin=actual_bin, observed_result=verdict, program_result=trial['status'], note=note, screenshot=trial['screenshot'])
            fresh = not path.exists()
            with path.open('a', newline='', encoding='utf-8-sig') as stream:
                writer = csv.DictWriter(stream, fieldnames=list(row))
                if fresh:
                    writer.writeheader()
                writer.writerow(row)
        self.notify('人工观察记录已保存；程序执行完毕不自动算作抓取成功。')

    def vision_worker(self):
        capture = cv2.VideoCapture(self.config['camera_device'])
        try:
            if not capture.isOpened():
                raise RuntimeError('摄像头无法打开，检查占用和编号')
            while not self.camera_stop.is_set():
                ok, raw = capture.read()
                if not ok or raw is None:
                    raise RuntimeError('摄像头取帧失败；已取消新任务并请求中断')
                with self.lock:
                    roi = copy.deepcopy(self.roi)
                result = detect_color(raw, self.config, roi)
                with self.lock:
                    self.last = dict(raw=raw.copy(), annotated=result['annotated'].copy(), result=result)
                    self.last_time = time.monotonic()
                    target = self.gate.feed(result) if roi is not None else None
                    state = self.gate.state
                    count = self.gate.count
                    mode = 'hardware' if self.backend.hardware else 'simulation'
                    stage = self.stage
                    if target:
                        self.jobs.put(('sort', dict(target=target, raw=raw.copy(), annotated=result['annotated'].copy())))
                display = result['annotated'].copy()
                cv2.putText(display, '{} {} stable={} {}'.format(mode, state, count, stage), (10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 2)
                self.image_callback(display)
                self.camera_stop.wait(self.config['vision']['frame_delay_seconds'])
        except Exception as error:
            self.stop()
            with self.lock:
                self.gate.error = True
                self.gate.state = 'ERROR'
            self.log('camera_error', error=str(error))
            self.notify(str(error))
        finally:
            capture.release()

    def motion_worker(self):
        while True:
            op, data = self.jobs.get()
            if op == 'shutdown':
                self.backend.close()
                self.backend.hardware = False
                with self.lock:
                    self.gate.state = 'CLOSED'
                return
            started = time.monotonic()
            status = 'done'
            error = None
            trial = None
            try:
                if op == 'connect':
                    if data:
                        precheck(self.config, self.config['sdk_dir'])
                    self.backend.close()
                    self.backend = Backend(self.config, self.cancel, hardware=data)
                    self.observation_ready = not data
                    self.roi = None
                    self.notify('真机已连接，未移动；点击回观察位。' if data else '已切换演练模式。')
                elif op == 'observe':
                    self.backend.move(self.config['observation_pose'])
                    self.observation_ready = True
                    self.roi = None
                    self.notify('已回观察位；请在当前画面重新设置 P4 ROI。')
                elif op == 'sort':
                    target = data['target']
                    stamp = datetime.now().strftime('%Y%m%d_%H%M%S_%f')
                    image = self.save_image(data['annotated'], stamp + '_detect.png')
                    self.save_image(data['raw'], stamp + '_raw.png')
                    trial = dict(trial_id=stamp, color=target['color'], mode='hardware' if self.backend.hardware else 'simulation', screenshot=image)
                    self.log('sort_start', **trial, target=target, plan_color=target['color'])

                    def stage(name):
                        with self.lock:
                            self.stage = name
                        self.log('stage', trial_id=stamp, stage=name)
                    self.backend.execute(target['color'], stage)
                elif op == 'hold':
                    self.backend.hold()
            except Cancelled:
                status = 'interrupted'
            except Exception as exc:
                status = 'error'
                error = str(exc)
                self.cancel.set()
                self.log('exception', operation=op, error=error, traceback=traceback.format_exc())
            if self.cancel.is_set() and op not in ('hold', 'connect'):
                try:
                    self.log('hold_result', result=self.backend.hold())
                except Exception as exc:
                    status = 'error'
                    error = (error + '; ' if error else '') + '制停失败: ' + str(exc)
            with self.lock:
                if self.backend.hardware and (status != 'done' or op == 'hold'):
                    self.observation_ready = False
                    self.roi = None
                if trial:
                    trial.update(status=status, error=error, elapsed_seconds=round(time.monotonic() - started, 3))
                    self.latest_trial = trial
                    self.log('sort_result', **trial)
                    self.gate.finish(status == 'done')
                    if status == 'interrupted':
                        self.gate.error = False
                        self.gate.stop()
                    self.notify('{} → {}：{}；移走/清空 P4 等待重置。'.format(trial['color'], self.config['bin_map'][trial['color']]['label'], status))
                else:
                    self.gate.busy = False
                    self.gate.error = bool(error)
                    self.gate.state = 'ERROR' if error else 'STOPPED'
                self.stage = ''
            if error:
                self.notify(error)

    def close(self):
        if self.closing:
            return
        self.closing = True
        self.camera_stop.set()
        self.stop()
        self.jobs.put(('shutdown', None))
        self.notify('已关闭预览并请求制停；待线程结束后可关闭 notebook 内核。')

def create_panel(config=None):
    import ipywidgets as widgets
    from IPython.display import display
    config = copy.deepcopy(CONFIG) if config is None else config
    try:
        previous = globals().get('_runtime')
        if previous and previous.camera_thread and previous.camera_thread.is_alive():
            raise RuntimeError('旧预览仍在运行；先停止旧面板，等待线程退出，再重建')
        if previous and (previous.gate.busy or previous.backend.hardware):
            raise RuntimeError('旧机械臂连接仍存在；请先停止并重启内核，不能重复连接')
    except NameError:
        pass
    status = widgets.HTML(value='默认演练；在 Jetson 本机摄像头预览，先框定 P4。')
    runtime = Runtime(config, lambda text: setattr(status, 'value', str(text)))
    globals()['_runtime'] = runtime
    image = widgets.Image(format='jpeg', width=640)

    def show(frame):
        ok, encoded = cv2.imencode('.jpg', frame)
        if ok:
            image.value = encoded.tobytes()
    runtime.image_callback = show

    def guard(fn):

        def handler(_):
            try:
                if runtime.closing:
                    raise RuntimeError('面板正在关闭；等待线程结束后重新创建面板')
                fn()
            except Exception as error:
                status.value = str(error)
        return handler
    buttons = []
    for label, fn in [('启动摄像头', runtime.start_camera), ('开始自动分拣', runtime.start_auto), ('停止当前动作', runtime.stop), ('解除停止', runtime.clear_stop), ('保存截图', runtime.snapshot), ('关闭预览', runtime.close)]:
        button = widgets.Button(description=label)
        button.on_click(guard(fn))
        buttons.append(button)
    hardware = widgets.Button(description='连接真机')
    hardware.on_click(guard(lambda: runtime.connect(True)))
    simulate = widgets.Button(description='切回演练')
    simulate.on_click(guard(lambda: runtime.connect(False)))
    observe = widgets.Button(description='回观察位')
    observe.on_click(guard(runtime.observe))
    roi = [widgets.IntText(value=v, description=k, layout=widgets.Layout(width='145px')) for k, v in zip(('x', 'y', 'width', 'height'), (200, 240, 240, 200))]
    roi_button = widgets.Button(description='应用 P4 ROI')
    roi_button.on_click(guard(lambda: runtime.apply_roi([w.value for w in roi])))
    actual = widgets.Dropdown(options=[(c, c) for c in COLORS], description='真实颜色')
    actual_bin = widgets.Dropdown(options=['厨余垃圾', '其他垃圾', '可回收物', '有害垃圾', '未投放/掉落'], description='实际投放')
    verdict = widgets.Dropdown(options=['成功', '失败', '未触发', '误触发'], description='人工判定')
    note = widgets.Text(description='现场备注')
    record = widgets.Button(description='记录本轮实测')
    record.on_click(guard(lambda: runtime.record_trial(actual.value, actual_bin.value, verdict.value, note.value)))
    display(widgets.VBox([widgets.HTML(value='<b>Ch6 固定 P4 四色自动分拣</b>'), widgets.HBox(buttons[:3]), widgets.HBox(buttons[3:]), widgets.HBox([hardware, simulate, observe]), image, widgets.HBox(roi + [roi_button]), status, widgets.HBox([actual, actual_bin, verdict]), widgets.HBox([note, record])]))
    return runtime
```

```python
def read_teaching_pose(label='green'):
    # Optional manual call only. No Home/open/move/damping command is sent.
    # Close other programs that own the arm UART before calling this function.
    robot = InterruptibleArm(CONFIG['sdk_dir'], CONFIG['arm_port'], threading.Event())
    try:
        x,y,z,pitch,roll = [float(v) for v in robot.arm.get_tool_pose()]
        if not all(math.isfinite(v) for v in (x,y,z,pitch,roll)):
            raise ValueError('Invalid pose feedback')
        k,b = robot.arm.z_error_polyfit
        command_pose = dict(x=x, y=y, z=z-float(k)*math.hypot(x,y)-float(b),
                            pitch_deg=math.degrees(pitch), roll_deg=math.degrees(roll))
        print(label, command_pose)
        print('Command pose: radians converted once; SDK Z correction removed for replay.')
        return command_pose
    finally:
        robot.close()

# Optional, after teaching and releasing the other UART connection:
# taught_green = read_teaching_pose('green')
# Copy the confirmed measurements into CONFIG in the first code cell.
```

```python
def offline_self_test():
    palette = {'red': (0,0,255), 'yellow': (0,255,255), 'green': (0,180,0), 'blue': (255,0,0)}
    for color, bgr in palette.items():
        frame = np.full((240,320,3),200,np.uint8)
        cv2.rectangle(frame,(80,80),(140,140),bgr,-1)
        found = detect_color(frame,CONFIG,[60,60,140,140])
        assert found['kind']=='target' and found['target']['color']==color
    gate=SortGate(frames=3,empty_frames=2,reset_delay=0)
    gate.start()
    target={'kind':'target','target':{'color':'red','center':(100,100)}}
    assert gate.feed(target) is None
    assert gate.feed(target) is None
    assert gate.feed(target)['color']=='red'
    for _ in range(10):
        assert gate.feed({'kind':'empty'}) is None
    gate.finish(True)
    for _ in range(10):
        assert gate.feed(target) is None
    assert gate.need_empty
    gate.feed({'kind':'empty'})
    gate.feed({'kind':'empty'})
    assert not gate.need_empty
    print('Offline self-test passed: four colors, debounce, BUSY and no repeated pickup.')
offline_self_test()
```

## 7 操作界面

1. 点击启动摄像头，按真实画面设置 P4 的 x/y/width/height，点击应用 P4 ROI。不要把彩色桶放进 ROI。
2. 点击开始自动分拣，一次只在 P4 放一块。稳定 6 帧后触发，无需选择颜色。
3. 演练不会拿走物块，需手动移走；动作结束后连续 8 帧空区确认才重新准备。运动遮挡不会导致重复抓取。
4. 真机：完成第 1 节参数确认，先关闭旧面板，重新运行参数单元，再运行本单元。点击连接真机→回观察位→重新设 P4 ROI→开始自动分拣。关闭其他占用串口的程序。
5. 点击停止当前动作会请求取消轨迹并保持位置，不代替物理急停。故障时不自动开爪或回零。
6. 结束时点击关闭预览，等待串口和线程退出。

结果自动保存到当前 notebook 目录的 results/时间/。每轮结束后填写真实颜色、实际桶和人工判定并记录。至少做 8 次随机实测，模拟结果不能写成真机成功。
