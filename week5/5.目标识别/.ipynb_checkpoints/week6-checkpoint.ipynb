# 实验六｜基于颜色识别的垃圾桶分拣

按实验报告完成：四色 HSV 检测、连续帧去抖、颜色到四桶的映射、`READY → DETECT → CONFIRM → BUSY → RESET → READY`、防重复抓取、真实截图与 8 次人工核实的随机测试记录。

**默认真机模式。** 启动预览只会打开摄像头；识别稳定后点击“分拣一次”才会执行机械臂动作。抓取期间摄像头继续实时显示识别画面，但不会再触发分拣。首次点击分拣前需核对四桶位姿、取物位姿、手眼标定、机械臂串口及运动空间。现有桶位数据沿用原 notebook，尚未在本次修改中实测，不能把它们当作已验证结果。Notebook 停止按钮只阻止后续任务，运动中的紧急停止须用设备急停。

```python
# 现场调参区：只在这里调整摄像头、ROI、HSV、确认帧数和机械臂参数
import csv, math, sys, threading, time, traceback
from pathlib import Path
from datetime import datetime
from dataclasses import dataclass
import cv2
import numpy as np
import ipywidgets as widgets
from IPython.display import display

SDK_DIR = Path('/home/jetson/Notebook/4.机械臂OpenCV视觉处理实验')
BASE = SDK_DIR
if str(SDK_DIR) not in sys.path:
    sys.path.insert(0, str(SDK_DIR))
CAMERA_ID = 0
DRY_RUN = False           # 默认执行真机；只有点击“分拣一次”才会发机械臂指令
ROI = None              # None 为整幅画面；也可填 (x1, y1, x2, y2)
MIN_AREA = 700
CONFIRM_FRAMES = 6
POSITION_TOLERANCE = 18
EMPTY_REARM_FRAMES = 8
FRAME_DELAY = 0.04
SAFE_DZ = 55.0
MOVE_T = 2.0
GRIPPER_T = 1.0
PICK_PITCH_DEG = 114.5
PICK_ROLL_DEG = 144.5
PICK_Z_OFFSET_WS = 5.0
HSV_RANGES = {
    'red': [((0, 80, 50), (10, 255, 255)), ((170, 80, 50), (179, 255, 255))],
    'yellow': [((18, 80, 60), (38, 255, 255))],
    'green': [((40, 60, 35), (90, 255, 255))],
    'blue': [((90, 75, 45), (135, 255, 255))],
}
BGR = {'red': (0, 0, 255), 'yellow': (0, 255, 255),
       'green': (0, 255, 0), 'blue': (255, 0, 0)}
CN = {'red': '红', 'yellow': '黄', 'green': '绿', 'blue': '蓝'}
KERNEL = np.ones((5, 5), np.uint8)
print('课程目录：', BASE)
print('当前模式：', '模拟' if DRY_RUN else '真机')
```

## 1. 颜色—垃圾桶映射与目标位姿

桶坐标来自原 `week6.ipynb`，单位假定为机械臂基坐标系的 mm，角度为度。原 Roll 数值超出单圈，下面保留原值并换算到 `[-180°, 180°]`；换算只表示等效方向，**不能代替现场示教、逆解与避障验证**。报告 4.1 可直接抄写运行后显示的映射表，并写明实际标定备注。

```python
@dataclass(frozen=True)
class Pose:
    x: float
    y: float
    z: float
    pitch_deg: float
    roll_deg: float
    def xyz(self):
        return [self.x, self.y, self.z]

RAW_ROLL = {'green': -945200.4, 'yellow': 137329.9,
            'blue': 137329.9, 'red': 148789.0}

def normalize_roll(degrees):
    """把原始角度换算为单圈的等效工具角度。"""
    return round((degrees + 180.0) % 360.0 - 180.0, 1)

bin_map = {
    'red':    {'bin': '有害垃圾', 'xyz': (127.4, -131.8, 146.4), 'pitch_deg': 94.5},
    'yellow': {'bin': '其他垃圾', 'xyz': (219.0, 99.8, 135.6), 'pitch_deg': 106.5},
    'green':  {'bin': '厨余垃圾', 'xyz': (122.8, 114.8, 122.8), 'pitch_deg': 112.7},
    'blue':   {'bin': '可回收物', 'xyz': (207.8, -141.3, 133.3), 'pitch_deg': 104.5},
}
for color, info in bin_map.items():
    info['roll_deg'] = normalize_roll(RAW_ROLL[color])
    print(f'{CN[color]} → {info["bin"]}，XYZ={info["xyz"]} mm，'
          f'Pitch={info["pitch_deg"]}°，Roll={info["roll_deg"]}°，'
          f'原始 Roll={RAW_ROLL[color]}°（待现场复核）')
```

## 2. HSV 检测与离线自检

先把图像转 HSV，按颜色生成 Mask，再开闭运算、按面积过滤轮廓。红色保留跨 H=0/179 的两段阈值。检测接口返回所有有效目标；分拣时只有画面中**恰好一个**有效目标才会进入连续帧确认，避免最大轮廓恰好选错。

```python
def detect_color(frame, colors=None, roi=None):
    """返回有效目标列表、标注画面和四色掩码。"""
    if frame is None or frame.size == 0:
        raise ValueError('输入画面为空')
    chosen = list(bin_map) if colors is None else list(colors)
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    height, width = frame.shape[:2]
    x1, y1, x2, y2 = (0, 0, width, height) if roi is None else roi
    x1, y1 = max(0, int(x1)), max(0, int(y1))
    x2, y2 = min(width, int(x2)), min(height, int(y2))
    if x1 >= x2 or y1 >= y2:
        raise ValueError('ROI 越界或为空')
    spatial = np.zeros((height, width), dtype=np.uint8)
    spatial[y1:y2, x1:x2] = 255
    canvas = frame.copy()
    detections, masks = [], {}
    for color in bin_map:
        mask = np.zeros((height, width), dtype=np.uint8)
        for lower, upper in HSV_RANGES[color]:
            mask |= cv2.inRange(hsv, np.array(lower, np.uint8), np.array(upper, np.uint8))
        mask = cv2.bitwise_and(mask, spatial)
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, KERNEL)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, KERNEL)
        masks[color] = mask
        if color not in chosen:
            continue
        contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        for contour in contours:
            area = cv2.contourArea(contour)
            if area < MIN_AREA:
                continue
            x, y, w, h = cv2.boundingRect(contour)
            target = {'color': color, 'pixel': (x + w // 2, y + h // 2),
                      'box': (x, y, w, h), 'area': float(area)}
            detections.append(target)
            cv2.rectangle(canvas, (x, y), (x + w, y + h), BGR[color], 2)
            cv2.putText(canvas, f'{color} {area:.0f}', (max(5, x), max(25, y - 8)),
                        cv2.FONT_HERSHEY_SIMPLEX, .6, BGR[color], 2)
    if roi is not None:
        cv2.rectangle(canvas, (x1, y1), (x2, y2), (220, 220, 220), 1)
    return detections, canvas, masks

def mask_montage(masks):
    """把四色 Mask 排成一张便于截图的图。"""
    tiles = []
    for color in bin_map:
        tile = cv2.cvtColor(masks[color], cv2.COLOR_GRAY2BGR)
        cv2.putText(tile, color, (12, 28), cv2.FONT_HERSHEY_SIMPLEX,
                    .8, BGR[color], 2)
        tiles.append(tile)
    return cv2.vconcat([cv2.hconcat(tiles[:2]), cv2.hconcat(tiles[2:])])

# 离线图像只用于算法检查，绝不作为实验报告的实拍证据。
sample = np.zeros((480, 640, 3), np.uint8)
for color, x in zip(bin_map, (35, 185, 335, 485)):
    cv2.rectangle(sample, (x, 175), (x + 90, 265), BGR[color], -1)
found, _, _ = detect_color(sample)
assert {item['color'] for item in found} == set(bin_map), '四色离线自检失败'
print('四色 HSV 离线自检通过；这不是实机测试结果。')
```

## 3. 抓取与放置接口

预览使用最简 `cv2.VideoCapture(CAMERA_ID)`，不强制设置分辨率、帧率或编码。真机启动预览时载入课程标定；**实拍分辨率须与标定映射尺寸一致**，检测使用去畸变画面并把该画面的像素交给 `Image2Workspace`。真机运动前先检查抓取、抬升、投放与退出点的逆解。预览线程独占摄像头，动作线程独占机械臂。BUSY 期间继续读帧、识别和显示，禁止接收新的分拣请求。

```python
def checked_bin(color):
    """检查桶位数值，并生成机械臂目标姿态。"""
    info = bin_map[color]
    numbers = (*info['xyz'], info['pitch_deg'], info['roll_deg'])
    if not np.all(np.isfinite(numbers)) or not -180 <= info['roll_deg'] <= 180:
        raise ValueError(f'{CN[color]}色桶位参数非法')
    return Pose(*info['xyz'], info['pitch_deg'], info['roll_deg'])

def above(pose):
    """生成目标点正上方的安全点。"""
    return Pose(pose.x, pose.y, pose.z + SAFE_DZ, pose.pitch_deg, pose.roll_deg)

def prepare_mapping(frame_shape):
    """仅载入现有标定文件，不再打开第二个摄像头。"""
    from camera import Camera
    from image2workspace import Image2Workspace
    from arm_application import ArmApplication
    import yaml
    cam = Camera.__new__(Camera)
    cam.config_folder = str(SDK_DIR / 'vconfig')
    with (SDK_DIR / 'vconfig' / 'camera.yaml').open(encoding='utf-8') as stream:
        cam.config = yaml.safe_load(stream)
    if cam.config['roi']['enable']:
        raise ValueError('课程相机配置启用了裁剪 ROI；请先重做标定或关闭该裁剪')
    cam.load_cam_calib_data()
    if tuple(cam.remap_x.shape[:2]) != tuple(frame_shape[:2]):
        raise ValueError(f'摄像头实际尺寸 {frame_shape[1]}×{frame_shape[0]} 与标定尺寸 '
                         f'{cam.remap_x.shape[1]}×{cam.remap_x.shape[0]} 不一致，禁止真机')
    mapper = Image2Workspace(cam, config_folder=str(SDK_DIR / 'vconfig'))
    return cam, mapper, ArmApplication

def precheck_poses(arm, poses):
    """不发运动指令，检查所有目标点是否能求出逆解。"""
    for name, pose in poses:
        adjusted = arm.adjust_tool_posi(pose.xyz())
        ok, _ = arm.inverse_kinematic(adjusted, math.radians(pose.pitch_deg),
                                      math.radians(pose.roll_deg))
        if not ok:
            raise ValueError(f'{name} 逆解不可达；未执行抓取')

def move_to(arm, pose, name):
    """按已核查的位姿执行一段阻塞运动。"""
    result = arm.set_tool_pose(pose.xyz(), tool_pitch=math.radians(pose.pitch_deg),
                               tool_roll=math.radians(pose.roll_deg), T=MOVE_T,
                               is_soft_move=True, is_wait=True)
    if result is False:
        raise RuntimeError(f'{name} 逆解失败，运动已停止')

def run_pick_place(target, mapper=None, app_class=None, cam_calib=None, dry_run=True):
    """执行一次抓取与投放；模拟模式只返回计划，不连接舵机。"""
    color, pixel = target['color'], target['pixel']
    destination = checked_bin(color)
    if dry_run:
        return f'模拟：{CN[color]}色 {pixel} → {bin_map[color]["bin"]}，未执行机械臂动作'
    if DRY_RUN or mapper is None or app_class is None or cam_calib is None:
        raise RuntimeError('真机坐标映射未就绪，禁止发出机械臂指令')
    wx, wy = mapper.img2workspace(*pixel)
    wz = float(mapper.object_height) + PICK_Z_OFFSET_WS
    if not np.all(np.isfinite([wx, wy, wz])):
        raise ValueError('手眼标定换算结果非法')
    from arm5dof_uservo import Arm5DoFUServo
    arm = Arm5DoFUServo(is_init_pose=False, config_folder=str(SDK_DIR / 'config'))
    try:
        return _run_hardware_sequence(arm, app_class, wx, wy, wz, color, destination)
    finally:
        arm.uart.close()

def _run_hardware_sequence(arm, app_class, wx, wy, wz, color, destination):
    """在串口保持打开期间完成单次抓放动作。"""
    app = app_class(arm, config_folder=str(SDK_DIR / 'config'))
    ax, ay, az = app.tf_ws2arm(wx, wy, wz)
    pick = Pose(float(ax), float(ay), float(az), PICK_PITCH_DEG, PICK_ROLL_DEG)
    if pick.z < 10 or above(pick).z > 350:
        raise ValueError('抓取高度超出保守范围')
    poses = [('抓取上方', above(pick)), ('抓取点', pick),
             ('桶上方', above(destination)), ('桶位', destination)]
    precheck_poses(arm, poses)
    arm.gripper_open(T=GRIPPER_T, is_wait=True)
    move_to(arm, above(pick), '抓取上方')
    move_to(arm, pick, '抓取点')
    arm.gripper_close(T=GRIPPER_T, is_wait=True)
    move_to(arm, above(pick), '抓取抬升')
    arm.home()
    move_to(arm, above(destination), '桶上方')
    move_to(arm, destination, '桶位')
    arm.gripper_open(T=GRIPPER_T, is_wait=True)
    move_to(arm, above(destination), '退出桶位')
    arm.home()
    return f'真机指令已结束：{CN[color]}色 → {bin_map[color]["bin"]}；实际投放仍需人工观察'
```

## 4. 实时预览与状态互锁

按顺序运行到此单元。启动预览后直接进入 `READY`；画面中可预先放好**一个**物块。识别连续稳定 6 帧后显示 `CONFIRM`，点击“分拣一次”。完成后必须把旧物块移出画面并保持 8 帧，系统才重新进入 `READY`。重新启动预览会直接识别当前物块，但仍需手动点击“分拣一次”；每次动作后的空场确认用于防止同一物体重复抓取。抓取期间可继续看到实时识别画面并按“保存实拍截图”取证；BUSY 期间不会重复抓取。保存截图只保存当前真实摄像头帧，不能把上方合成图作为报告证据。

```python
# 重新运行界面单元时，先释放上一次的预览线程。
if 'runtime' in globals():
    old_job = runtime.get('job_thread')
    if old_job is not None and old_job.is_alive():
        raise RuntimeError('机械臂动作仍在执行；请等待完成后再运行本单元')
    old_thread = runtime.get('thread')
    if old_thread is not None and old_thread.is_alive():
        runtime['stop'].set()
        old_thread.join(timeout=3)
        if old_thread.is_alive():
            raise RuntimeError('旧预览线程仍占用摄像头；请等待旧任务结束后再运行本单元')

color_choice = widgets.Dropdown(options=[('四色自动识别', 'all'),
    ('只识别红色', 'red'), ('只识别黄色', 'yellow'),
    ('只识别绿色', 'green'), ('只识别蓝色', 'blue')],
    value='all', description='颜色')
btn_start = widgets.Button(description='启动预览', button_style='success')
btn_sort = widgets.Button(description='分拣一次', button_style='warning')
btn_snapshot = widgets.Button(description='保存实拍截图', button_style='info')
btn_stop = widgets.Button(description='停止预览', button_style='danger')
state_label = widgets.HTML(value='状态：STOPPED')
image_widget = widgets.Image(format='jpeg', width=640)
mask_widget = widgets.Image(format='jpeg', width=640)
log_box = widgets.Output(layout={'border': '1px solid #ddd', 'height': '160px', 'overflow': 'auto'})
display(widgets.VBox([widgets.HBox([color_choice, btn_start, btn_sort, btn_snapshot, btn_stop]),
    state_label, widgets.HTML('<b>原画标注</b>'), image_widget,
    widgets.HTML('<b>四色 Mask</b>'), mask_widget, log_box]))

ACTION_CSV = BASE / 'week6_action_log.csv'
TEST_CSV = BASE / 'week6_test_records.csv'
EVIDENCE_DIR = BASE / 'evidence_week6'
lock = threading.RLock()
if ACTION_CSV.exists():
    with ACTION_CSV.open(newline='', encoding='utf-8-sig') as stream:
        previous_job_count = sum(1 for _ in csv.DictReader(stream))
else:
    previous_job_count = 0
runtime = {'thread': None, 'stop': threading.Event(), 'state': 'STOPPED',
           'candidate': None, 'stable': None, 'count': 0, 'empty': 0, 'frame_number': 0,
           'pending': False, 'latest': None, 'last_job': None,
           'job_number': previous_job_count, 'recorded_jobs': set(), 'mode': None,
           'job_thread': None}

def ui_log(message):
    with log_box:
        print(datetime.now().strftime('%H:%M:%S'), message)

def set_state(state, detail=''):
    with lock:
        runtime['state'] = state
        count = runtime['count']
    state_label.value = f'<b>状态：{state}</b>　稳定帧 {count}/{CONFIRM_FRAMES}　{detail}'

def append_csv(path, fields, row):
    """逐条追加记录，保留此前真实测试记录。"""
    exists = path.exists() and path.stat().st_size > 0
    with path.open('a', newline='', encoding='utf-8-sig') as stream:
        writer = csv.DictWriter(stream, fieldnames=fields)
        if not exists:
            writer.writeheader()
        writer.writerow(row)

def matching_target(a, b):
    """比较前后帧的颜色与中心，避免把不同物块计为连续帧。"""
    return bool(a and b and a['color'] == b['color']
                and abs(a['pixel'][0] - b['pixel'][0]) <= POSITION_TOLERANCE
                and abs(a['pixel'][1] - b['pixel'][1]) <= POSITION_TOLERANCE)

def run_job(target, mode, mapper, app_class, cam_calib):
    """独立执行一次抓放；预览线程继续识别和显示，但不得触发新任务。"""
    error = ''
    try:
        message = run_pick_place(target, mapper, app_class, cam_calib, dry_run=mode)
        ui_log(message)
    except Exception as exc:
        error = str(exc)
        ui_log('动作失败：' + error)
        with log_box:
            print(traceback.format_exc())
    with lock:
        runtime['job_number'] += 1
        job = {'编号': runtime['job_number'], '时间': datetime.now().isoformat(timespec='seconds'),
               '识别颜色': target['color'], '目标桶': bin_map[target['color']]['bin'],
               '像素坐标': str(target['pixel']), '模式': '模拟' if mode else '真机',
               '执行状态': '失败' if error else '指令结束，待人工核实', '异常': error}
        runtime['last_job'] = job
        runtime.update(candidate=None, stable=None, count=0, empty=0)
        if runtime['state'] == 'ERROR':
            pass
        elif error:
            set_state('ERROR', '动作失败；禁止新任务，请检查硬件')
        elif runtime['stop'].is_set():
            set_state('STOPPED', '动作已结束，摄像头已停止')
        else:
            set_state('RESET', '移走旧物块，等待空场确认')
    append_csv(ACTION_CSV, list(job), job)

def preview_loop(mode):
    capture = None
    try:
        capture = cv2.VideoCapture(CAMERA_ID)
        if not capture.isOpened():
            raise RuntimeError(f'摄像头 {CAMERA_ID} 打开失败，请检查设备号与占用')
        cam_calib = mapper = app_class = None
        if not mode:
            ok, first = capture.read()
            if not ok:
                raise RuntimeError('摄像头首次取帧失败')
            cam_calib, mapper, app_class = prepare_mapping(first.shape)
        with lock:
            runtime['empty'] = 0
            runtime['count'] = 0
        set_state('READY', '等待单个目标连续稳定后可点击分拣')
        previous_choice = color_choice.value
        while not runtime['stop'].is_set():
            ok, raw = capture.read()
            if not ok:
                raise RuntimeError('摄像头取帧失败，请检查摄像头连接或占用')
            if cam_calib and cam_calib.config['vflip']:
                raw = cv2.flip(raw, 0)
            if cam_calib and cam_calib.config['hflip']:
                raw = cv2.flip(raw, 1)
            frame = cam_calib.remove_distortion(raw) if cam_calib else raw
            choice = color_choice.value
            if choice != previous_choice:
                with lock:
                    runtime.update(candidate=None, stable=None, count=0, pending=False)
                previous_choice = choice
            colors = None if choice == 'all' else [choice]
            detections, canvas, masks = detect_color(frame, colors, ROI)
            # 即使只选一种颜色，重新武装也要求整幅画面没有四类目标。
            all_detections = detections if choice == 'all' else detect_color(frame, None, ROI)[0]
            single = detections[0] if len(detections) == 1 and len(all_detections) == 1 else None
            with lock:
                state = runtime['state']
                if state == 'RESET':
                    runtime['empty'] = runtime['empty'] + 1 if not all_detections else 0
                    if runtime['empty'] >= EMPTY_REARM_FRAMES:
                        runtime.update(candidate=None, stable=None, count=0, pending=False)
                        set_state('READY', '可放入一个物块')
                elif state in ('READY', 'DETECT', 'CONFIRM'):
                    prior = runtime['candidate']
                    runtime['count'] = min(CONFIRM_FRAMES, runtime['count'] + 1) if matching_target(prior, single) else (1 if single else 0)
                    runtime['candidate'] = single
                    runtime['stable'] = dict(single) if single and runtime['count'] >= CONFIRM_FRAMES else None
                    set_state('CONFIRM' if runtime['stable'] else ('DETECT' if all_detections else 'READY'),
                              '仅允许一个目标' if len(all_detections) > 1 else '')
                pending = runtime['pending'] and runtime['state'] == 'CONFIRM'
                target = dict(runtime['stable']) if pending and runtime['stable'] else None
                if target:
                    runtime['pending'] = False
                    set_state('BUSY', '持续显示识别画面；禁止再次分拣')
            with lock:
                runtime['frame_number'] += 1
                frame_text = f'{runtime["state"]}  frame {runtime["frame_number"]}'
            cv2.putText(canvas, frame_text, (12, 28), cv2.FONT_HERSHEY_SIMPLEX,
                        .8, (255, 255, 255), 2)
            montage = mask_montage(masks)
            with lock:
                runtime['latest'] = (raw.copy(), canvas.copy(), montage.copy())
            encoded, jpeg = cv2.imencode('.jpg', canvas, [cv2.IMWRITE_JPEG_QUALITY, 80])
            if encoded:
                image_widget.value = jpeg.tobytes()
            encoded, jpeg = cv2.imencode('.jpg', montage, [cv2.IMWRITE_JPEG_QUALITY, 70])
            if encoded:
                mask_widget.value = jpeg.tobytes()
            if target:
                job_thread = threading.Thread(target=run_job,
                    args=(target, mode, mapper, app_class, cam_calib), daemon=True)
                with lock:
                    runtime['job_thread'] = job_thread
                job_thread.start()
            time.sleep(FRAME_DELAY)
    except Exception as exc:
        ui_log('预览失败：' + str(exc))
        set_state('ERROR', str(exc))
    finally:
        if capture is not None:
            capture.release()
        with lock:
            runtime['pending'] = False
            if runtime['state'] != 'ERROR':
                set_state('STOPPED', '摄像头已释放')

def on_start(_):
    with lock:
        if runtime['thread'] is not None and runtime['thread'].is_alive():
            return
        if runtime['job_thread'] is not None and runtime['job_thread'].is_alive():
            ui_log('上一次机械臂动作仍在执行，请等待结束')
            return
        runtime['stop'] = threading.Event()
        runtime.update(candidate=None, stable=None, count=0, pending=False, latest=None, frame_number=0,
                       mode='模拟' if DRY_RUN else '真机')
        runtime['thread'] = threading.Thread(target=preview_loop, args=(bool(DRY_RUN),), daemon=True)
        runtime['thread'].start()
    ui_log('启动预览：' + runtime['mode'] + '；单个目标稳定后可点击分拣')

def on_sort(_):
    with lock:
        if runtime['state'] != 'CONFIRM' or runtime['stable'] is None:
            ui_log(f'当前状态 {runtime["state"]}：请等待目标稳定；若刚完成分拣，须先移走旧物块')
            return
        runtime['pending'] = True
    ui_log('已请求分拣一次')

def on_snapshot(_):
    with lock:
        latest = runtime['latest']
        if latest is None:
            ui_log('当前没有可保存的实拍画面')
            return
        pictures = tuple(image.copy() for image in latest)
    EVIDENCE_DIR.mkdir(exist_ok=True)
    stamp = datetime.now().strftime('%Y%m%d_%H%M%S_%f')
    for label, picture in zip(('原图', '识别', '四色Mask'), pictures):
        cv2.imwrite(str(EVIDENCE_DIR / f'{stamp}_{label}.png'), picture)
    ui_log(f'已保存真实摄像头画面：{EVIDENCE_DIR / stamp}')

def on_stop(_):
    runtime['stop'].set()
    with lock:
        runtime['pending'] = False
    ui_log('停止请求已发出；若机械臂正在运动，须等待动作完成或使用设备急停')

btn_start.on_click(on_start)
btn_sort.on_click(on_sort)
btn_snapshot.on_click(on_snapshot)
btn_stop.on_click(on_stop)
```

## 5. 8 次随机测试与人工核实记录

每次**真机**分拣指令结束后，看实际物块颜色和实际投放桶，再选择下方两项并点击“记录本次实测”。成功条件由“识别颜色等于实际颜色、实际桶等于目标桶、动作没有报错”自动计算。模拟执行不会计入 8 次实测。即使投放失败，也选择“未进入桶”并写异常。每次动作只允许记录一次；CSV 逐行追加，不覆盖旧记录。

```python
actual_color = widgets.Dropdown(options=[('请选择', ''), *[(CN[c], c) for c in bin_map]],
                                description='实际颜色')
actual_bin = widgets.Dropdown(options=[('请选择', ''),
    ('未进入桶', '未进入桶'), *[(info['bin'], info['bin']) for info in bin_map.values()]],
    description='实际投放桶')
duplicate_pick = widgets.Dropdown(options=[('请选择', ''), ('否', '否'), ('是', '是')],
                                  description='重复抓取')
test_note = widgets.Text(description='异常/备注', placeholder='例如阴影、偏抓或无异常')
btn_record = widgets.Button(description='记录本次实测', button_style='info')
test_output = widgets.Output()
display(widgets.VBox([actual_color, actual_bin, duplicate_pick, test_note, btn_record, test_output]))

def on_record(_):
    with lock:
        job = runtime['last_job']
        if job is None or job['模式'] != '真机':
            with test_output:
                print('目前没有待核实的真机动作；模拟结果不能计入 8 次测试。')
            return
        if job['编号'] in runtime['recorded_jobs']:
            with test_output:
                print('本次动作已记录，不能重复登记。')
            return
        if not actual_color.value or not actual_bin.value or not duplicate_pick.value:
            with test_output:
                print('请先选择实际颜色、实际投放桶和是否重复抓取。')
            return
        success = (job['执行状态'] != '失败' and
                   job['识别颜色'] == actual_color.value and
                   job['目标桶'] == actual_bin.value and duplicate_pick.value == '否')
        row = {'序号': job['编号'], '时间': job['时间'], '物块实际颜色': CN[actual_color.value],
               '识别颜色': CN[job['识别颜色']], '目标桶': job['目标桶'],
               '实际投放桶': actual_bin.value, '成功/失败': '成功' if success else '失败',
               '异常/备注': test_note.value or job['异常'] or '无', '重复抓取': duplicate_pick.value}
        append_csv(TEST_CSV, list(row), row)
        runtime['recorded_jobs'].add(job['编号'])
    with test_output:
        print('已保存实测记录：', row, '文件：', TEST_CSV)

btn_record.on_click(on_record)
```

## 6. 报告取证与现场步骤

1. 先依次执行本 notebook 的代码单元。默认 `DRY_RUN=False`，启动预览即为真机模式；画面中可预先放好一个彩块。稳定识别后点击“分拣一次”，并按“保存实拍截图”记录真实画面。
2. 对照原课程标定文件与当前实体布置，核实四个桶的 XYZ、Pitch、Roll、取物高度、上方点和回零路径。原 notebook 中的桶 Z 与课程 `config.py` 的示例不同，必须以现场实测为准；同时检查串口与舵机 ID。运动前程序会检查映射尺寸和各目标点逆解。
3. 每次只放一个彩块，等待 `CONFIRM` 后点击“分拣一次”。抓取时预览画面继续识别和显示；在抓取、投放阶段点击“保存实拍截图”。若机械臂遮住目标，可从旁边用实际相机补拍，不能使用模拟图片充当证据。
4. 每次动作后在第 5 节选择实际颜色、实际投放桶，填写异常并记录。移走旧物块，等待 `RESET → READY`，再乱序放下一个。至少 8 次。报告中的成功次数、成功率与重复抓取情况从真实记录和现场观察填写。

`week6_action_log.csv` 是动作日志，`week6_test_records.csv` 是人工核实的实测记录，截图位于 `evidence_week6/`。更改 HSV 或位姿后，在报告中写下更改前后的参数、问题现象和再次验证结果；程序不会编造这些数据。

```python
if TEST_CSV.exists():
    with TEST_CSV.open(newline='', encoding='utf-8-sig') as stream:
        rows = list(csv.DictReader(stream))
    real_rows = rows[:8]
    correct = sum(row['成功/失败'] == '成功' for row in real_rows)
    repeated = any(row['重复抓取'] == '是' for row in real_rows)
    print('已记录真实测试：', len(rows), '次；前 8 次正确：', correct, '/ 8' if len(rows) >= 8 else '（未满 8 次）')
    if len(rows) >= 8:
        print('前 8 次成功率：', f'{correct / 8 * 100:.1f}%', '；重复抓取：', '是' if repeated else '否', '；验收至少 6/8 且无重复抓取')
    for row in rows:
        print(row)
else:
    print('尚无人工核实的真机测试记录。')
```
