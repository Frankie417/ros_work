# 实验六｜基于 HSV 颜色识别的垃圾分类分拣（交互界面）

**颜色 → 固定桶位**：🟢 绿色 → 厨余垃圾；🟡 黄色 → 其他垃圾；🔵 蓝色 → 可回收物；🔴 红色 → 有害垃圾。

工作流：`READY → DETECT → CONFIRM → BUSY → RESET → READY`。可在界面选择“只抓选定颜色”或“全部颜色自动分拣”，点击 **请求分拣一次** 后才能触发；连续帧验证、目标位置验证、BUSY 锁和需移除目标的重新武装机制共同降低重复抓取风险。

**安全默认值：模拟模式（不会驱动机械臂）**。实际动作必须先完成相机与九点手眼标定、确认四个垃圾桶位姿、确认舵机通讯正常，并在场地清空后勾选实际执行。程序不包含未经确认的真实机械臂坐标。

在 Jetson 原实验目录运行本 Notebook；同目录应有 `camera.py`、`arm5dof_uservo.py`、`arm_application.py`、`image2workspace.py`、`config/`、`vconfig/`。支持 Jupyter 的 `ipywidgets`，视频通过小部件更新（无须 MobaXterm OpenCV GUI）。

python3 -m pip install opencv-python numpy ipywidgets ipython pyyaml

原实验的 `ArmApplication.grab_cubic(wx,wy,wz,color_name)` 会根据颜色前往 `object_green/object_yellow/object_blue/object_red`，并用 `object_*_add` 退出，然后返回 `capture_image`。**这些命名位姿必须在原机械臂配置中完成实体标定，不可直接猜测坐标。**

````python
from pathlib import Path
import os, time, threading, csv, traceback
from datetime import datetime
from collections import deque
import cv2, numpy as np
import ipywidgets as widgets
from IPython.display import display

BASE = Path.cwd()
CAMERA_ID = 0
FRAME_DELAY = 0.08
CONFIRM_FRAMES = 8             # 连续确认帧数
MIN_AREA = 700                # 像素轮廓面积阈值
MIN_WH, MAX_WH = 35, 350      # 原实验 100-300 像素可现场调整
POSITION_TOLERANCE = 18      # 连续帧的中心位置允许漂移（像素）
ABSENT_FRAMES_TO_REARM = 10  # 成功执行后要求目标离开画面
ROI = None                   # (x1,y1,x2,y2)，None 为全画幅
ARM_OFFSET_XY = -10.0        # 沿径向修正，单位 mm；来自原实验，务必现场标定
ARM_OFFSET_Z = 0.0           # 单位 mm
PICK_X_BIAS = -25.0          # 原实验抓取修正，单位 mm；务必现场标定
PICK_Y_BIAS = 10.0

BIN_MAP = {
    'green':  {'zh': '厨余垃圾', 'pose': 'object_green',  'after_pose': 'object_green_add'},
    'yellow': {'zh': '其他垃圾', 'pose': 'object_yellow', 'after_pose': 'object_yellow_add'},
    'blue':   {'zh': '可回收物', 'pose': 'object_blue',   'after_pose': 'object_blue_add'},
    'red':    {'zh': '有害垃圾', 'pose': 'object_red',    'after_pose': 'object_red_add'},
}
HSV_RANGES = {  # 源自 real_week5.ipynb 风格；请根据真实光照校正
    'red': [((0, 90, 55), (10, 255, 255)), ((170, 90, 55), (179, 255, 255))],
    'yellow': [((18, 75, 60), (38, 255, 255))],
    'green': [((40, 60, 35), (90, 255, 255))],
    'blue': [((95, 70, 45), (135, 255, 255))],
}
BGR = {'red': (0, 0, 255), 'yellow': (0, 255, 255),
       'green': (0, 255, 0), 'blue': (255, 0, 0)}
print('工作目录:', BASE)
print('桶位映射:', BIN_MAP)
````

## 2. 检测算法：HSV → Mask → 形态学 → 轮廓 → 方框

与第五周 HSV 实验保持一致。不使用原压缩包内的 `ColorBlockDetector`，因为它用 BGR 多元高斯建模而非 HSV；机械臂驱动仍复用原实验接口。

````python
def detect_color(frame, colors=None):
    colors = colors or list(BIN_MAP)
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    kernel = np.ones((5, 5), np.uint8)
    canvas = frame.copy()
    H, W = frame.shape[:2]
    x1, y1, x2, y2 = ROI if ROI else (0, 0, W, H)
    x1, y1 = max(0, x1), max(0, y1)
    x2, y2 = min(W, x2), min(H, y2)
    detections, masks = [], {}
    for color in colors:
        mask = np.zeros((H, W), dtype=np.uint8)
        for lo, hi in HSV_RANGES[color]:
            mask = cv2.bitwise_or(mask, cv2.inRange(hsv, np.array(lo, np.uint8), np.array(hi, np.uint8)))
        roi_mask = np.zeros_like(mask)
        roi_mask[y1:y2, x1:x2] = 255
        mask = cv2.bitwise_and(mask, roi_mask)
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
        masks[color] = mask
        contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        for cnt in contours:
            area = cv2.contourArea(cnt)
            x, y, w, h = cv2.boundingRect(cnt)
            if area < MIN_AREA or not (MIN_WH <= w <= MAX_WH and MIN_WH <= h <= MAX_WH):
                continue
            cx, cy = x + w // 2, y + h // 2
            detections.append(dict(color=color, cx=cx, cy=cy, bbox=(x,y,w,h), area=area))
            cv2.rectangle(canvas, (x,y), (x+w,y+h), BGR[color], 2)
            cv2.circle(canvas, (cx,cy), 4, BGR[color], -1)
            cv2.putText(canvas, f'{color} {int(area)}', (x,max(20,y-8)), cv2.FONT_HERSHEY_SIMPLEX, .6, BGR[color], 2)
    if ROI:
        cv2.rectangle(canvas, (x1,y1), (x2,y2), (200,200,200), 1)
    return detections, canvas, masks

# 无需硬件的离线自检
sample = np.zeros((480, 640, 3), dtype=np.uint8)
for c, x in zip(BIN_MAP, [45, 190, 335, 480]):
    cv2.rectangle(sample, (x,185), (x+90,280), BGR[c], -1)
found, rendered, _ = detect_color(sample)
print('离线检测到:', [(d['color'], d['cx'], d['cy']) for d in found])
assert set(d['color'] for d in found) == set(BIN_MAP), 'HSV 自检未通过，请检查阈值'
````

## 3. 机械臂桥接（默认禁用实际动作）

仅实际执行时加载原实验的 `Camera`、`Image2Workspace`、`Arm5DoFUServo`、`ArmApplication`。必须先确认 `KeyError: 0` 等舵机注册故障已经排除。`ArmApplication.grab_cubic` 本身负责抓、抬、按颜色投放并返回拍照位。

````python
def validate_hardware_files():
    paths = ['camera.py', 'image2workspace.py', 'arm5dof_uservo.py',
             'arm_application.py', 'config/T_arm2ws.txt', 'config/object.yaml',
             'vconfig/handeye_9points.yaml']
    missing = [p for p in paths if not (BASE / p).exists()]
    if missing:
        raise FileNotFoundError('缺少实验依赖: ' + ', '.join(missing))
    # 原实验部分位姿配置在驱动模块中实现，无法仅据文件名保证已标定。
    return True

hardware = None

def prepare_hardware():
    global hardware
    if hardware is not None:
        return hardware
    validate_hardware_files()
    from camera import Camera
    from image2workspace import Image2Workspace
    from arm5dof_uservo import Arm5DoFUServo
    from arm_application import ArmApplication
    # 机械臂构造函数可能立即 home()，初始化前必须有人监护并清空工作区。
    arm = Arm5DoFUServo(config_folder='config')
    app = ArmApplication(arm, config_folder='config')
    cam_for_mapping = Camera()
    cam_for_mapping.load_cam_calib_data()
    mapper = Image2Workspace(cam_for_mapping)
    hardware = (arm, app, mapper, cam_for_mapping)
    return hardware

def do_sort(target, live=False):
    color, px, py = target['color'], target['cx'], target['cy']
    dest = BIN_MAP[color]
    if not live:
        return f'[模拟] {color} 像素({px},{py}) → {dest["zh"]} / {dest["pose"]}'
    arm, app, mapper, _ = prepare_hardware()
    wx, wy = mapper.img2workspace(px, py)
    if not (np.isfinite(wx) and np.isfinite(wy)):
        raise ValueError('手眼标定坐标异常')
    # 径向修正沿用原实验
    theta = np.arctan2(wy, wx)
    r = np.hypot(wx, wy) + ARM_OFFSET_XY
    wx2 = r*np.cos(theta) + PICK_X_BIAS
    wy2 = r*np.sin(theta) + PICK_Y_BIAS
    wz2 = mapper.object_height + ARM_OFFSET_Z
    app.grab_cubic(wx2, wy2, wz2, color)
    return f'[实机] {color} ({wx2:.1f},{wy2:.1f},{wz2:.1f})mm → {dest["zh"]}'
````

## 4. 交互界面：选颜色、看实时画面、请求抓取

**操作顺序**：启动预览 → 选择颜色 → 等待“稳定目标” → 点击“请求分拣一次”。默认仅模拟，日志仍会记录目标桶位。实机执行前必须勾选安全确认、切换模式，并确保桶位命名位姿已验证。为避免重复抓取，执行后同一目标需离开检测区至少指定帧数，才允许再次请求。

````python
selection = widgets.Dropdown(options=[('全部颜色自动匹配', 'all'), ('绿色 → 厨余', 'green'),
    ('黄色 → 其他', 'yellow'), ('蓝色 → 可回收', 'blue'), ('红色 → 有害', 'red')],
    value='all', description='目标颜色')
mode = widgets.ToggleButtons(options=[('模拟分拣','sim'), ('真实机械臂','live')], value='sim', description='模式')
ack = widgets.Checkbox(value=False, description='已完成机械臂/桶位标定、清空危险区域并有人员监护')
start_btn = widgets.Button(description='启动预览', button_style='success')
stop_btn = widgets.Button(description='停止预览', button_style='warning')
request_btn = widgets.Button(description='请求分拣一次', button_style='info')
reset_btn = widgets.Button(description='重新允许分拣（确认已移除目标）')
state_html = widgets.HTML(value='<b>状态：</b>STOPPED')
frame_view = widgets.Image(format='jpeg', width=640, height=480)
log_out = widgets.Output(layout={'max_height':'180px','overflow':'auto'})

runtime = {'running':False, 'state':'STOPPED', 'candidate':None,
           'count':0, 'stable':None, 'request':False, 'thread':None,
           'absent':0, 'last_target':None, 'reset_after_job':False}
lock = threading.RLock()
CSV_PATH = BASE / 'week6_sorting_log.csv'

def ui_log(message):
    with log_out:
        print(datetime.now().strftime('%H:%M:%S'), message)

def mark_state(s):
    runtime['state'] = s
    state_html.value = f'<b>状态：</b>{s} ｜ 稳定帧：{runtime["count"]}/{CONFIRM_FRAMES}'

def save_attempt(target, status, note):
    exists = CSV_PATH.exists()
    with CSV_PATH.open('a', newline='', encoding='utf-8-sig') as f:
        writer = csv.writer(f)
        if not exists:
            writer.writerow(['time','color','garbage','target_pose','pixel_x','pixel_y','status','note'])
        c = target['color']
        writer.writerow([datetime.now().isoformat(timespec='seconds'), c, BIN_MAP[c]['zh'],
                         BIN_MAP[c]['pose'], target['cx'], target['cy'], status, note])

def run_job(target, live):
    try:
        msg = do_sort(target, live=live)
        save_attempt(target, '执行完成(未人工验证投放)', msg)
        ui_log(msg)
        with lock:
            runtime['last_target'] = target
            runtime['absent'] = 0
            runtime['request'] = False
            mark_state('RESET / 等待目标移除')
    except Exception as e:
        ui_log('执行失败：'+ repr(e))
        save_attempt(target, '失败', repr(e))
        with lock:
            runtime['request'] = False
            mark_state('ERROR / 检查硬件后手动复位')

def preview_loop():
    cam = cv2.VideoCapture(CAMERA_ID)
    if not cam.isOpened():
        ui_log('摄像头无法打开，请检查 CAMERA_ID 和摄像头占用')
        with lock:
            runtime['running'] = False
            mark_state('STOPPED')
        return
    with lock:
        mark_state('READY')
    try:
        while runtime['running']:
            ret, frame = cam.read()
            if not ret:
                ui_log('摄像头读帧失败')
                break
            colors = list(BIN_MAP) if selection.value == 'all' else [selection.value]
            detections, canvas, _ = detect_color(frame, colors)
            # 多目标时使用最大面积目标；稳定性需要连续帧的颜色和位置一致。
            candidate = max(detections, key=lambda d:d['area']) if detections else None
            with lock:
                s = runtime['state']
                if s in ('READY','DETECT','CONFIRM'):
                    prev = runtime['candidate']
                    same = (candidate and prev and candidate['color']==prev['color']
                            and abs(candidate['cx']-prev['cx']) <= POSITION_TOLERANCE
                            and abs(candidate['cy']-prev['cy']) <= POSITION_TOLERANCE)
                    runtime['count'] = runtime['count']+1 if same else (1 if candidate else 0)
                    runtime['candidate'] = candidate
                    if runtime['count'] >= CONFIRM_FRAMES:
                        runtime['stable'] = dict(candidate)
                        mark_state('CONFIRM')
                    else:
                        runtime['stable'] = None
                        mark_state('DETECT' if candidate else 'READY')
                    if runtime['request'] and runtime['stable'] is not None:
                        tgt = dict(runtime['stable'])
                        runtime['request'] = False
                        mark_state('BUSY')
                        threading.Thread(target=run_job, args=(tgt, mode.value=='live'), daemon=True).start()
                elif s.startswith('RESET'):
                    t = runtime['last_target']
                    still = any(d['color']==t['color'] and
                                abs(d['cx']-t['cx']) < POSITION_TOLERANCE*2 and
                                abs(d['cy']-t['cy']) < POSITION_TOLERANCE*2 for d in detections) if t else False
                    runtime['absent'] = 0 if still else runtime['absent'] + 1
                    if runtime['absent'] >= ABSENT_FRAMES_TO_REARM:
                        runtime['count']=0; runtime['candidate']=None; runtime['stable']=None
                        mark_state('READY')
                label = runtime['state']
            cv2.putText(canvas, label[:35], (12, 25), cv2.FONT_HERSHEY_SIMPLEX, .7, (255,255,255), 2)
            ok, jpg = cv2.imencode('.jpg', canvas, [cv2.IMWRITE_JPEG_QUALITY, 75])
            if ok:
                frame_view.value = jpg.tobytes()
            time.sleep(FRAME_DELAY)
    finally:
        cam.release()
        with lock:
            runtime['running'] = False
            if runtime['state'] not in ('BUSY',):
                mark_state('STOPPED')

def on_start(_):
    with lock:
        if runtime['running'] or runtime['state']=='BUSY': return
        runtime.update(running=True, request=False, count=0, candidate=None, stable=None, last_target=None)
        t = threading.Thread(target=preview_loop, daemon=True)
        runtime['thread'] = t
        t.start()

def on_stop(_):
    with lock:
        runtime['running']=False
        runtime['request']=False
    ui_log('预览停止指令已发出；此操作不构成机械臂紧急停止。')

def on_request(_):
    with lock:
        if runtime['state'] != 'CONFIRM' or not runtime['stable']:
            ui_log('尚未找到稳定目标 / 系统非 READY，拒绝分拣')
            return
        if mode.value=='live' and not ack.value:
            ui_log('实机模式必须完成安全确认')
            return
        runtime['request']=True
        ui_log(f'请求一次：{runtime["stable"]["color"]} → {BIN_MAP[runtime["stable"]["color"]]["zh"]}')

def on_reset(_):
    with lock:
        if runtime['state'] in ('BUSY',):
            ui_log('BUSY 期间禁止复位'); return
        if runtime['state'].startswith('RESET') or runtime['state'].startswith('ERROR'):
            runtime['count']=0; runtime['candidate']=None; runtime['stable']=None
            runtime['last_target']=None; runtime['request']=False
            mark_state('READY' if runtime['running'] else 'STOPPED')
            ui_log('手动复位；请确认已移除旧物块、处理好硬件异常')

start_btn.on_click(on_start); stop_btn.on_click(on_stop)
request_btn.on_click(on_request); reset_btn.on_click(on_reset)
display(widgets.VBox([
    widgets.HTML('<h3>四色垃圾分拣控制台</h3>'),
    widgets.HBox([selection, mode]), ack,
    widgets.HBox([start_btn, stop_btn, request_btn]), reset_btn, state_html,
    frame_view, widgets.HTML('<b>操作日志</b>'), log_out
]))
````
