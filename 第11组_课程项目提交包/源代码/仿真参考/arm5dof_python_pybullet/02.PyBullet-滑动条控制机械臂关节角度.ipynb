```python
import time
import numpy as np
from numpy import pi
import pybullet
import pybullet_data
# 自定义库
from arm5dof_pybullet import Arm5DoFPyBullet

# 设置Numpy的打印选项
# 精确位数3，不启用科学计数法
np.set_printoptions(precision=5, suppress=True)
```

## 创建机械臂对象

```python
# 创建机械臂
# 传入机械臂路径
arm = Arm5DoFPyBullet("urdf/arm5dof.urdf")
```

## GUI组建

```python
def create_gui_sliders():
	'''滑动条 GUI '''
	global arm
	sliders = []   # 滑动条列表
	
	for joint_name in arm.joint_name_list[:-1]:
		# 创建用户调试参数，并添加到滑动条列表里面
		lowerb = arm.joint_info_dict[joint_name].lower_limit # 关节角度下限
		upperb = arm.joint_info_dict[joint_name].upper_limit # 关节角度上限
		init_value = 0 # 关节角度初始值
		slider = pybullet.addUserDebugParameter(" "+joint_name, lowerb, upperb, init_value)
		sliders.append(slider)
    
	return sliders

def read_gui_sliders(sliders):
	'''GUI滑动条 数据读取'''
	theta1 = pybullet.readUserDebugParameter(sliders[0])
	theta2 = pybullet.readUserDebugParameter(sliders[1])
	theta3 = pybullet.readUserDebugParameter(sliders[2])
	theta4 = pybullet.readUserDebugParameter(sliders[3])
	theta5 = pybullet.readUserDebugParameter(sliders[4])
	theta_gripper = pybullet.readUserDebugParameter(sliders[5])
	return [theta1, theta2, theta3, theta4, theta5, theta_gripper]
```

```python
# 创建滑动条
sliders = create_gui_sliders()
```

```python
# 读取滑动条的数值
read_gui_sliders(sliders)
```

```python

while True:
    # 更新关节角度
    theta_angles = read_gui_sliders(sliders)
    # 设置关节角度
    arm.set_joint_angle_list(theta_angles)
```
